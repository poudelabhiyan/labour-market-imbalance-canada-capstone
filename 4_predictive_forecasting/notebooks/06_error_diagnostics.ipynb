```python
# ============================================================
# 06_error_diagnostics.ipynb
# MAIN CELL — PHASE C: ERROR DIAGNOSTICS
# ============================================================
# Purpose
# -------
# This notebook analyzes rolling-origin out-of-sample forecast
# errors from the 4 frozen base models:
#   1. SARIMAX
#   2. TBATS-Prophet
#   3. VAR
#   4. BiLSTM-Attention
#
# Input files
# -----------
#   exports/backtesting/rolling_predictions_unemployment.csv
#   exports/backtesting/rolling_predictions_vacancies.csv
#   exports/backtesting/rolling_predictions_theta.csv
#
# Output goals
# ------------
# 1. Error correlation matrix across models
# 2. Error performance by forecast horizon
# 3. Forecast bias by model
# 4. Turning-point / directional accuracy
# 5. Error distribution diagnostics
# 6. Exportable thesis-ready tables
#
# Why this phase matters
# ----------------------
# Ensemble learning is justified only if models make partly
# complementary errors. This notebook tests that directly using
# honest rolling-origin out-of-sample predictions.
# ============================================================

# ============================================================
# 0. IMPORTS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# 1. SAFE DISPLAY FALLBACK
# ============================================================

try:
    display
except NameError:
    def display(x):
        print(x)

# ============================================================
# 2. CONFIGURATION
# ============================================================

BACKTEST_DIR = Path("exports/backtesting")
DIAG_DIR = Path("exports/error_diagnostics")
DIAG_DIR.mkdir(parents=True, exist_ok=True)

TARGET_FILES = {
    "unemployment": BACKTEST_DIR / "rolling_predictions_unemployment.csv",
    "vacancies": BACKTEST_DIR / "rolling_predictions_vacancies.csv",
    "theta": BACKTEST_DIR / "rolling_predictions_theta.csv",
}

MODEL_COLS = {
    "SARIMAX": "sarimax_pred",
    "TBATS_PROPHET": "tbats_prophet_pred",
    "VAR": "var_pred",
    "BiLSTM_Attention": "bilstm_pred",
}

# Regime cut points: edit if your thesis uses a different partition
# These are only for diagnostics, not for model fitting.
def assign_regime(date_value):
    d = pd.to_datetime(date_value)
    if d < pd.Timestamp("2020-03-01"):
        return "Pre-COVID"
    elif d <= pd.Timestamp("2021-12-01"):
        return "COVID Shock"
    elif d <= pd.Timestamp("2023-12-01"):
        return "Labour Tightness"
    else:
        return "Recent Period"

# ============================================================
# 3. HELPER FUNCTIONS
# ============================================================

def clean_pair(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    mask = (
        ~np.isnan(y_true) &
        ~np.isnan(y_pred) &
        ~np.isinf(y_true) &
        ~np.isinf(y_pred)
    )
    return y_true[mask], y_pred[mask]

def rmse(y_true, y_pred):
    y_true, y_pred = clean_pair(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return np.sqrt(np.mean((y_true - y_pred) ** 2))

def mae(y_true, y_pred):
    y_true, y_pred = clean_pair(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return np.mean(np.abs(y_true - y_pred))

def mape(y_true, y_pred):
    y_true, y_pred = clean_pair(y_true, y_pred)
    mask = y_true != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

def smape(y_true, y_pred):
    y_true, y_pred = clean_pair(y_true, y_pred)
    denom = np.abs(y_true) + np.abs(y_pred)
    mask = denom != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(2 * np.abs(y_pred[mask] - y_true[mask]) / denom[mask]) * 100

def mase(y_true, y_pred):
    y_true, y_pred = clean_pair(y_true, y_pred)
    if len(y_true) < 2:
        return np.nan
    naive = np.mean(np.abs(np.diff(y_true)))
    if naive == 0:
        return np.nan
    return np.mean(np.abs(y_true - y_pred)) / naive

def directional_accuracy(actual_series, pred_series):
    """
    Compare sign of first differences between actual and predicted.
    Uses sorted forecast_date order.
    """
    actual = pd.Series(actual_series).astype(float).reset_index(drop=True)
    pred = pd.Series(pred_series).astype(float).reset_index(drop=True)

    actual_diff = actual.diff()
    pred_diff = pred.diff()

    mask = (
        ~actual_diff.isna() &
        ~pred_diff.isna() &
        ~np.isinf(actual_diff) &
        ~np.isinf(pred_diff)
    )

    if mask.sum() == 0:
        return np.nan

    actual_sign = np.sign(actual_diff[mask])
    pred_sign = np.sign(pred_diff[mask])

    return np.mean(actual_sign == pred_sign)

def pivot_error_matrix(long_df):
    """
    Build a wide error matrix with one column per model for correlation analysis.
    Index keys: origin, forecast_date, horizon
    """
    wide = long_df.pivot_table(
        index=["origin", "forecast_date", "horizon"],
        columns="model",
        values="error",
        aggfunc="first"
    )
    return wide

# ============================================================
# 4. LOAD ROLLING PREDICTIONS
# ============================================================

target_dfs = {}

for target, path in TARGET_FILES.items():
    if not path.exists():
        raise FileNotFoundError(
            f"Missing required file: {path}\n"
            f"Run 05_rolling_origin_backtesting.ipynb first."
        )

    temp = pd.read_csv(path)
    temp["origin"] = pd.to_datetime(temp["origin"])
    temp["forecast_date"] = pd.to_datetime(temp["forecast_date"])
    if "train_end_date" in temp.columns:
        temp["train_end_date"] = pd.to_datetime(temp["train_end_date"])

    target_dfs[target] = temp.copy()

print("Loaded rolling backtesting files successfully.\n")
for k, v in target_dfs.items():
    print(f"{k}: {v.shape}")

# ============================================================
# 5. CONVERT TO LONG ERROR TABLES
# ============================================================

long_error_tables = {}

for target, df_target in target_dfs.items():
    long_rows = []

    for model_name, pred_col in MODEL_COLS.items():
        temp = df_target[["origin", "forecast_date", "horizon", "actual", pred_col]].copy()
        temp = temp.rename(columns={pred_col: "prediction"})
        temp["model"] = model_name
        temp["target"] = target
        temp["error"] = temp["actual"] - temp["prediction"]
        temp["abs_error"] = np.abs(temp["error"])
        temp["sq_error"] = temp["error"] ** 2
        temp["ape"] = np.where(temp["actual"] != 0, np.abs(temp["error"] / temp["actual"]) * 100, np.nan)
        temp["regime"] = temp["forecast_date"].apply(assign_regime)
        long_rows.append(temp)

    long_df = pd.concat(long_rows, ignore_index=True)
    long_error_tables[target] = long_df

print("\nLong error table preview — unemployment")
display(long_error_tables["unemployment"].head(10))

# ============================================================
# 6. MODEL-LEVEL ERROR SUMMARY
# ============================================================

model_summaries = {}

for target, long_df in long_error_tables.items():
    rows = []

    for model_name in long_df["model"].unique():
        temp = long_df[long_df["model"] == model_name].copy()

        rows.append({
            "target": target,
            "model": model_name,
            "n_obs": len(temp),
            "ME_bias": np.nanmean(temp["error"]),
            "MAE": mae(temp["actual"], temp["prediction"]),
            "RMSE": rmse(temp["actual"], temp["prediction"]),
            "MAPE": mape(temp["actual"], temp["prediction"]),
            "sMAPE": smape(temp["actual"], temp["prediction"]),
            "MASE": mase(temp["actual"], temp["prediction"]),
            "median_abs_error": np.nanmedian(temp["abs_error"]),
            "std_error": np.nanstd(temp["error"]),
        })

    summary_df = pd.DataFrame(rows).sort_values("RMSE").reset_index(drop=True)
    model_summaries[target] = summary_df

    out_path = DIAG_DIR / f"error_summary_by_model_{target}.csv"
    summary_df.to_csv(out_path, index=False)

    print(f"\nModel-level error summary — {target}")
    display(summary_df)

# ============================================================
# 7. HORIZON-WISE ERROR SUMMARY
# ============================================================

horizon_summaries = {}

for target, long_df in long_error_tables.items():
    rows = []

    for h in sorted(long_df["horizon"].unique()):
        h_df = long_df[long_df["horizon"] == h].copy()

        for model_name in h_df["model"].unique():
            temp = h_df[h_df["model"] == model_name].copy()

            rows.append({
                "target": target,
                "horizon": h,
                "model": model_name,
                "n_obs": len(temp),
                "ME_bias": np.nanmean(temp["error"]),
                "MAE": mae(temp["actual"], temp["prediction"]),
                "RMSE": rmse(temp["actual"], temp["prediction"]),
                "MAPE": mape(temp["actual"], temp["prediction"]),
                "sMAPE": smape(temp["actual"], temp["prediction"]),
                "MASE": mase(temp["actual"], temp["prediction"]),
            })

    horizon_df = pd.DataFrame(rows)
    horizon_summaries[target] = horizon_df

    out_path = DIAG_DIR / f"error_summary_by_horizon_{target}.csv"
    horizon_df.to_csv(out_path, index=False)

print("\nHorizon-wise error summary — unemployment preview")
display(horizon_summaries["unemployment"].head(12))

# ============================================================
# 8. REGIME-WISE ERROR SUMMARY
# ============================================================

regime_summaries = {}

for target, long_df in long_error_tables.items():
    rows = []

    for regime_name in sorted(long_df["regime"].dropna().unique()):
        r_df = long_df[long_df["regime"] == regime_name].copy()

        for model_name in r_df["model"].unique():
            temp = r_df[r_df["model"] == model_name].copy()

            rows.append({
                "target": target,
                "regime": regime_name,
                "model": model_name,
                "n_obs": len(temp),
                "ME_bias": np.nanmean(temp["error"]),
                "MAE": mae(temp["actual"], temp["prediction"]),
                "RMSE": rmse(temp["actual"], temp["prediction"]),
                "MAPE": mape(temp["actual"], temp["prediction"]),
                "sMAPE": smape(temp["actual"], temp["prediction"]),
                "MASE": mase(temp["actual"], temp["prediction"]),
            })

    regime_df = pd.DataFrame(rows)
    regime_summaries[target] = regime_df

    out_path = DIAG_DIR / f"error_summary_by_regime_{target}.csv"
    regime_df.to_csv(out_path, index=False)

print("\nRegime-wise error summary — unemployment")
display(regime_summaries["unemployment"])

# ============================================================
# 9. ERROR CORRELATION MATRICES
# ============================================================

error_correlation_tables = {}

for target, long_df in long_error_tables.items():
    error_wide = pivot_error_matrix(long_df)
    corr_df = error_wide.corr()
    error_correlation_tables[target] = corr_df

    out_path = DIAG_DIR / f"error_correlation_matrix_{target}.csv"
    corr_df.to_csv(out_path)

    print(f"\nError correlation matrix — {target}")
    display(corr_df)

# ============================================================
# 10. TURNING-POINT / DIRECTIONAL ACCURACY
# ============================================================

turning_point_tables = {}

for target, df_target in target_dfs.items():
    rows = []

    # Evaluate by horizon as well as overall
    for model_name, pred_col in MODEL_COLS.items():
        # Overall directional accuracy on horizon-1 sequence
        temp_h1 = df_target[df_target["horizon"] == 1].sort_values("forecast_date").copy()

        rows.append({
            "target": target,
            "scope": "overall_h1",
            "horizon": 1,
            "model": model_name,
            "directional_accuracy": directional_accuracy(temp_h1["actual"], temp_h1[pred_col]),
            "n_obs": len(temp_h1)
        })

        # Horizon-specific directional accuracy
        for h in sorted(df_target["horizon"].unique()):
            temp_h = df_target[df_target["horizon"] == h].sort_values("forecast_date").copy()

            rows.append({
                "target": target,
                "scope": "by_horizon",
                "horizon": h,
                "model": model_name,
                "directional_accuracy": directional_accuracy(temp_h["actual"], temp_h[pred_col]),
                "n_obs": len(temp_h)
            })

    tp_df = pd.DataFrame(rows)
    turning_point_tables[target] = tp_df

    out_path = DIAG_DIR / f"turning_point_accuracy_{target}.csv"
    tp_df.to_csv(out_path, index=False)

print("\nTurning-point accuracy — unemployment")
display(turning_point_tables["unemployment"].head(20))

# ============================================================
# 11. BIAS SUMMARY
# ============================================================

bias_tables = {}

for target, long_df in long_error_tables.items():
    rows = []

    for model_name in long_df["model"].unique():
        temp = long_df[long_df["model"] == model_name].copy()

        rows.append({
            "target": target,
            "model": model_name,
            "mean_error_bias": np.nanmean(temp["error"]),
            "median_error_bias": np.nanmedian(temp["error"]),
            "mean_signed_percentage_error": np.nanmean(
                np.where(temp["actual"] != 0, (temp["error"] / temp["actual"]) * 100, np.nan)
            ),
            "overprediction_share": np.mean(temp["error"] < 0),
            "underprediction_share": np.mean(temp["error"] > 0),
        })

    bias_df = pd.DataFrame(rows)
    bias_tables[target] = bias_df

    out_path = DIAG_DIR / f"error_bias_by_model_{target}.csv"
    bias_df.to_csv(out_path, index=False)

print("\nBias summary — unemployment")
display(bias_tables["unemployment"])

# ============================================================
# 12. VISUALS — ERROR CORRELATION HEATMAPS
# ============================================================

for target, corr_df in error_correlation_tables.items():
    plt.figure(figsize=(7, 6))
    plt.imshow(corr_df.values, aspect="auto")
    plt.xticks(range(len(corr_df.columns)), corr_df.columns, rotation=45, ha="right")
    plt.yticks(range(len(corr_df.index)), corr_df.index)
    plt.title(f"Error Correlation Matrix — {target}")
    plt.colorbar()
    plt.tight_layout()
    plt.show()

# ============================================================
# 13. VISUALS — RMSE BY HORIZON
# ============================================================

for target, h_df in horizon_summaries.items():
    plt.figure(figsize=(12, 6))

    for model_name in h_df["model"].unique():
        temp = h_df[h_df["model"] == model_name].sort_values("horizon")
        plt.plot(temp["horizon"], temp["RMSE"], marker="o", label=model_name)

    plt.title(f"RMSE by Forecast Horizon — {target}")
    plt.xlabel("Forecast Horizon")
    plt.ylabel("RMSE")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# 14. VISUALS — BIAS BY MODEL
# ============================================================

for target, bias_df in bias_tables.items():
    plt.figure(figsize=(10, 5))
    plt.bar(bias_df["model"], bias_df["mean_error_bias"])
    plt.title(f"Mean Forecast Error Bias by Model — {target}")
    plt.xlabel("Model")
    plt.ylabel("Mean Error (Actual - Forecast)")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# 15. VISUALS — TURNING-POINT ACCURACY (HORIZON 1)
# ============================================================

for target, tp_df in turning_point_tables.items():
    temp = tp_df[tp_df["scope"] == "overall_h1"].copy()

    plt.figure(figsize=(10, 5))
    plt.bar(temp["model"], temp["directional_accuracy"])
    plt.title(f"Turning-Point / Directional Accuracy — Horizon 1 — {target}")
    plt.xlabel("Model")
    plt.ylabel("Directional Accuracy")
    plt.ylim(0, 1)
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# 16. VISUALS — ERROR DISTRIBUTION HISTOGRAMS
# ============================================================

for target, long_df in long_error_tables.items():
    models = list(long_df["model"].unique())

    fig, axes = plt.subplots(len(models), 1, figsize=(12, 3 * len(models)), sharex=True)

    if len(models) == 1:
        axes = [axes]

    for ax, model_name in zip(axes, models):
        temp = long_df[long_df["model"] == model_name].copy()
        err = temp["error"].replace([np.inf, -np.inf], np.nan).dropna()

        ax.hist(err, bins=30)
        ax.axvline(0, linestyle="--")
        ax.set_title(f"Error Distribution — {target} — {model_name}")
        ax.set_ylabel("Count")
        ax.grid(True, alpha=0.3)

    axes[-1].set_xlabel("Forecast Error (Actual - Prediction)")
    plt.tight_layout()
    plt.show()

# ============================================================
# 17. THESIS-READY NARRATIVE SUMMARY TABLE
# ============================================================

narrative_rows = []

for target in ["unemployment", "vacancies", "theta"]:
    model_summary = model_summaries[target].sort_values("RMSE").reset_index(drop=True)
    bias_summary = bias_tables[target].copy()
    corr_df = error_correlation_tables[target].copy()

    best_model = model_summary.iloc[0]["model"]
    worst_model = model_summary.iloc[-1]["model"]

    # Average off-diagonal correlation
    corr_vals = corr_df.values.astype(float)
    if corr_vals.shape[0] > 1:
        off_diag = corr_vals[~np.eye(corr_vals.shape[0], dtype=bool)]
        avg_corr = np.nanmean(off_diag)
    else:
        avg_corr = np.nan

    # Best horizon-1 directional model
    tp_h1 = turning_point_tables[target]
    tp_h1 = tp_h1[tp_h1["scope"] == "overall_h1"].sort_values("directional_accuracy", ascending=False)

    best_tp_model = tp_h1.iloc[0]["model"] if len(tp_h1) > 0 else np.nan
    best_tp_score = tp_h1.iloc[0]["directional_accuracy"] if len(tp_h1) > 0 else np.nan

    narrative_rows.append({
        "target": target,
        "best_overall_model_by_rmse": best_model,
        "worst_overall_model_by_rmse": worst_model,
        "average_error_correlation": avg_corr,
        "best_h1_directional_model": best_tp_model,
        "best_h1_directional_accuracy": best_tp_score
    })

narrative_summary_df = pd.DataFrame(narrative_rows)
narrative_summary_df.to_csv(DIAG_DIR / "diagnostic_narrative_summary.csv", index=False)

print("\nDiagnostic narrative summary")
display(narrative_summary_df)

# ============================================================
# 18. FINAL INTERPRETATION BLOCK
# ============================================================

print("\n============================================================")
print("ERROR DIAGNOSTICS COMPLETED")
print("============================================================")
print("Files saved to:", DIAG_DIR.resolve())

print("\nCore exports created:")
print(" - error_summary_by_model_unemployment.csv")
print(" - error_summary_by_model_vacancies.csv")
print(" - error_summary_by_model_theta.csv")
print(" - error_summary_by_horizon_unemployment.csv")
print(" - error_summary_by_horizon_vacancies.csv")
print(" - error_summary_by_horizon_theta.csv")
print(" - error_summary_by_regime_unemployment.csv")
print(" - error_summary_by_regime_vacancies.csv")
print(" - error_summary_by_regime_theta.csv")
print(" - error_correlation_matrix_unemployment.csv")
print(" - error_correlation_matrix_vacancies.csv")
print(" - error_correlation_matrix_theta.csv")
print(" - turning_point_accuracy_unemployment.csv")
print(" - turning_point_accuracy_vacancies.csv")
print(" - turning_point_accuracy_theta.csv")
print(" - error_bias_by_model_unemployment.csv")
print(" - error_bias_by_model_vacancies.csv")
print(" - error_bias_by_model_theta.csv")
print(" - diagnostic_narrative_summary.csv")

```
