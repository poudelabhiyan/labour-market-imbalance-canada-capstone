```python
# ============================================================
# 05_rolling_origin_backtesting.ipynb
# MAIN CELL — PHASE B: ROLLING-ORIGIN BACKTESTING
# ============================================================
# Purpose
# -------
# This cell performs strict rolling-origin backtesting for the
# 4 frozen base models:
#   1. SARIMAX
#   2. TBATS-Prophet
#   3. VAR
#   4. BiLSTM-Attention
#
# Targets
# -------
#   - unemployment
#   - vacancies
#   - theta
#
# Outputs
# -------
# 1. rolling_predictions_unemployment.csv
# 2. rolling_predictions_vacancies.csv
# 3. rolling_predictions_theta.csv
# 4. model_comparison_unemployment.csv
# 5. model_comparison_vacancies.csv
# 6. model_comparison_theta.csv
#
# Design principles
# -----------------
# - Strictly out-of-sample forecasts only
# - Same rolling-origin windows for every model
# - Same metrics for every model
# - Same export contract for later ensemble training
# - Forecast-safe recursive logic for BiLSTM
#
# ============================================================

# ============================================================
# 0. IMPORTS
# ============================================================

import warnings
warnings.filterwarnings("ignore")

import os
import sys
import gc
import random
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import mean_squared_error, mean_absolute_error
from sklearn.preprocessing import StandardScaler

from statsmodels.tsa.statespace.sarimax import SARIMAX
from statsmodels.tsa.api import VAR

# TBATS / Prophet / TensorFlow imports
from tbats import TBATS
from prophet import Prophet

import tensorflow as tf
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, LSTM, Dense, Dropout, Bidirectional, Layer
from tensorflow.keras.callbacks import EarlyStopping
import tensorflow.keras.backend as K

# ============================================================
# 1. REPRODUCIBILITY
# ============================================================

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# ============================================================
# 2. SAFE DISPLAY FALLBACK
# ============================================================

try:
    display
except NameError:
    def display(x):
        print(x)

# ============================================================
# 3. CONFIGURATION
# ============================================================

# -------------------------------
# Backtesting window
# -------------------------------
BACKTEST_START = "2022-12-01"   # first origin date
MAX_HORIZON = 12                # 1 to 12 months ahead
ORIGIN_STEP = 1                 # move origin by 1 month each iteration

# -------------------------------
# Export directory
# -------------------------------
EXPORT_DIR = Path("exports/backtesting")
EXPORT_DIR.mkdir(parents=True, exist_ok=True)

# -------------------------------
# Frozen SARIMAX settings
# -------------------------------
SARIMAX_U_ORDER = (1, 1, 0)
SARIMAX_U_SEASONAL = (0, 0, 0, 12)

SARIMAX_V_ORDER = (2, 1, 1)
SARIMAX_V_SEASONAL = (1, 0, 1, 12)

# -------------------------------
# Frozen VAR settings
# -------------------------------
VAR_MAX_LAGS = 12
VAR_LAG_CRITERION = "aic"

# -------------------------------
# Frozen BiLSTM settings
# -------------------------------
LOOKBACK = 12
FEATURE_COLS = ["log_unemployment", "log_vacancies", "log_theta"]
TARGET_MAP = {
    "unemployment": "log_unemployment",
    "vacancies": "log_vacancies",
    "theta": "log_theta"
}
BILSTM_EPOCHS = 200
BILSTM_BATCH_SIZE = 16
BILSTM_DROPOUT = 0.2
BILSTM_PATIENCE = 10

# -------------------------------
# Optional runtime flags
# Keep all four True for full thesis run
# -------------------------------
RUN_SARIMAX = True
RUN_TBATS_PROPHET = True
RUN_VAR = True
RUN_BILSTM = True

# ============================================================
# 4. LOAD CANONICAL DATASET
# ============================================================

df = None

try:
    from data_loader import load_dataset
    df = load_dataset(verbose=False).copy()
except Exception:
    possible_paths = [
        Path.cwd(),
        Path.cwd().parent,
        Path.cwd() / "utils",
        Path.cwd().parent / "utils",
        Path.cwd() / "4_predictive_forecasting" / "utils",
        Path.cwd().parent / "4_predictive_forecasting" / "utils",
    ]

    loaded = False
    for p in possible_paths:
        if p.exists() and str(p) not in sys.path:
            sys.path.append(str(p))
        try:
            from data_loader import load_dataset
            df = load_dataset(verbose=False).copy()
            loaded = True
            break
        except Exception:
            continue

    if not loaded:
        raise ImportError(
            "Could not import load_dataset from data_loader.py. "
            "Make sure the loader is available."
        )

# Ensure datetime index
if "month" in df.columns:
    df["month"] = pd.to_datetime(df["month"])
    df = df.set_index("month")

df.index = pd.to_datetime(df.index)
df = df.sort_index()

required_cols = ["unemployment", "vacancies", "theta"]
missing_cols = [c for c in required_cols if c not in df.columns]
if missing_cols:
    raise ValueError(f"Missing required columns in dataset: {missing_cols}")

for c in required_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

if df[required_cols].isna().any().any():
    raise ValueError(
        "NaN values found in unemployment, vacancies, or theta. "
        "Clean the dataset before backtesting."
    )

# Add transformed columns once
df["log_unemployment"] = np.log1p(df["unemployment"])
df["log_vacancies"] = np.log1p(df["vacancies"])
df["log_theta"] = np.log1p(df["theta"])

print("Dataset loaded successfully.")
print("Shape:", df.shape)
print("Date range:", df.index.min(), "to", df.index.max())

# ============================================================
# 5. ROLLING-ORIGIN SETUP
# ============================================================

# We use actual index positions so forecast dates always align
all_dates = list(df.index)

# Backtest starts from the first index position >= BACKTEST_START
origin_start_idx = None
for i, d in enumerate(all_dates):
    if d >= pd.to_datetime(BACKTEST_START):
        origin_start_idx = i
        break

if origin_start_idx is None:
    raise ValueError("BACKTEST_START is after the available data range.")

# Last origin must leave room for MAX_HORIZON future actuals
origin_end_idx = len(all_dates) - MAX_HORIZON - 1

if origin_end_idx < origin_start_idx:
    raise ValueError("Not enough data for the chosen MAX_HORIZON and BACKTEST_START.")

origin_indices = list(range(origin_start_idx, origin_end_idx + 1, ORIGIN_STEP))
origin_dates = [all_dates[i] for i in origin_indices]

print("\nRolling-origin setup")
print("Number of origins:", len(origin_dates))
print("First origin:", origin_dates[0])
print("Last origin :", origin_dates[-1])
print("Max horizon :", MAX_HORIZON)

# ============================================================
# 6. METRIC FUNCTIONS
# ============================================================

def clean_metric_inputs(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mask = (
        ~np.isnan(y_true) &
        ~np.isnan(y_pred) &
        ~np.isinf(y_true) &
        ~np.isinf(y_pred)
    )
    return y_true[mask], y_pred[mask]

def safe_rmse(y_true, y_pred):
    y_true, y_pred = clean_metric_inputs(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return np.sqrt(mean_squared_error(y_true, y_pred))

def safe_mae(y_true, y_pred):
    y_true, y_pred = clean_metric_inputs(y_true, y_pred)
    if len(y_true) == 0:
        return np.nan
    return mean_absolute_error(y_true, y_pred)

def safe_mape(y_true, y_pred):
    y_true, y_pred = clean_metric_inputs(y_true, y_pred)
    mask = y_true != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y_true[mask] - y_pred[mask]) / y_true[mask])) * 100

def safe_smape(y_true, y_pred):
    y_true, y_pred = clean_metric_inputs(y_true, y_pred)
    denom = np.abs(y_true) + np.abs(y_pred)
    mask = denom != 0
    if mask.sum() == 0:
        return np.nan
    return np.mean(2 * np.abs(y_pred[mask] - y_true[mask]) / denom[mask]) * 100

def safe_mase(y_true, y_pred):
    y_true, y_pred = clean_metric_inputs(y_true, y_pred)
    if len(y_true) < 2:
        return np.nan
    naive_denom = np.mean(np.abs(np.diff(y_true)))
    if naive_denom == 0:
        return np.nan
    return np.mean(np.abs(y_true - y_pred)) / naive_denom

# ============================================================
# 7. BILSTM HELPER FUNCTIONS
# ============================================================

class AttentionLayer(Layer):
    def build(self, input_shape):
        self.W = self.add_weight(
            shape=(input_shape[-1], 1),
            initializer="random_normal",
            trainable=True
        )
        self.b = self.add_weight(
            shape=(input_shape[1], 1),
            initializer="zeros",
            trainable=True
        )
        super().build(input_shape)

    def call(self, x):
        e = K.tanh(K.dot(x, self.W) + self.b)
        a = K.softmax(e, axis=1)
        context = x * a
        context = K.sum(context, axis=1)
        return context

def build_sequences(dataframe, target_col, lookback):
    X = []
    y = []

    values = dataframe.values
    cols = list(dataframe.columns)
    target_idx = cols.index(target_col)

    for i in range(lookback, len(values)):
        X.append(values[i - lookback:i])
        y.append(values[i, target_idx])

    return np.array(X), np.array(y)

def build_bilstm_model(input_dim, lookback):
    inputs = Input(shape=(lookback, input_dim))

    x = Bidirectional(LSTM(64, return_sequences=True))(inputs)
    x = Dropout(BILSTM_DROPOUT)(x)

    x = Bidirectional(LSTM(32, return_sequences=True))(x)
    x = Dropout(BILSTM_DROPOUT)(x)

    x = AttentionLayer()(x)

    x = Dense(32, activation="relu")(x)
    x = Dropout(BILSTM_DROPOUT)(x)

    outputs = Dense(1)(x)

    model = Model(inputs, outputs)
    model.compile(optimizer="adam", loss="mse", metrics=["mae"])
    return model

def inverse_scaled_target(scaler, scaled_values, feature_cols, target_col):
    target_idx = feature_cols.index(target_col)

    dummy = np.zeros((len(scaled_values), len(feature_cols)))
    dummy[:, target_idx] = scaled_values
    inv = scaler.inverse_transform(dummy)
    return inv[:, target_idx]

def forecast_bilstm_recursive(train_df, horizon):
    """
    Train 3 separate frozen BiLSTM models:
      - unemployment
      - vacancies
      - theta

    Then recursively forecast the next horizon steps using only
    training information and previous predictions.
    """

    # Scale on training data only
    scaler = StandardScaler()
    train_scaled = pd.DataFrame(
        scaler.fit_transform(train_df[FEATURE_COLS]),
        index=train_df.index,
        columns=FEATURE_COLS
    )

    # Fit one model per target
    trained_models = {}

    for target_name, target_col in TARGET_MAP.items():
        X_train, y_train = build_sequences(train_scaled, target_col, LOOKBACK)

        if len(X_train) < 10:
            raise ValueError(
                f"Not enough sequences to train BiLSTM for {target_name}. "
                f"Need more history before origin {train_df.index.max()}."
            )

        # Validation split from training only
        val_size = max(1, int(len(X_train) * 0.15))
        X_train_core = X_train[:-val_size]
        y_train_core = y_train[:-val_size]
        X_val = X_train[-val_size:]
        y_val = y_train[-val_size:]

        if len(X_train_core) == 0:
            X_train_core = X_train
            y_train_core = y_train
            X_val = X_train[-1:]
            y_val = y_train[-1:]

        model = build_bilstm_model(input_dim=len(FEATURE_COLS), lookback=LOOKBACK)

        early_stop = EarlyStopping(
            monitor="val_loss",
            patience=BILSTM_PATIENCE,
            restore_best_weights=True
        )

        model.fit(
            X_train_core,
            y_train_core,
            validation_data=(X_val, y_val),
            epochs=BILSTM_EPOCHS,
            batch_size=BILSTM_BATCH_SIZE,
            callbacks=[early_stop],
            verbose=0
        )

        trained_models[target_name] = model

    # Recursive multivariate state update
    window = train_scaled.iloc[-LOOKBACK:].values.copy()

    pred_scaled_u = []
    pred_scaled_v = []
    pred_scaled_t = []

    for step in range(horizon):
        x_input = window[np.newaxis, :, :]

        next_u = trained_models["unemployment"].predict(x_input, verbose=0).flatten()[0]
        next_v = trained_models["vacancies"].predict(x_input, verbose=0).flatten()[0]
        next_t = trained_models["theta"].predict(x_input, verbose=0).flatten()[0]

        pred_scaled_u.append(next_u)
        pred_scaled_v.append(next_v)
        pred_scaled_t.append(next_t)

        next_row = np.array([next_u, next_v, next_t])
        window = np.vstack([window[1:], next_row])

    # Back-transform
    pred_log_u = inverse_scaled_target(scaler, np.array(pred_scaled_u), FEATURE_COLS, "log_unemployment")
    pred_log_v = inverse_scaled_target(scaler, np.array(pred_scaled_v), FEATURE_COLS, "log_vacancies")
    pred_log_t = inverse_scaled_target(scaler, np.array(pred_scaled_t), FEATURE_COLS, "log_theta")

    pred_u = np.expm1(pred_log_u)
    pred_v = np.expm1(pred_log_v)
    pred_t = np.expm1(pred_log_t)

    pred_u = np.clip(pred_u, 0, None)
    pred_v = np.clip(pred_v, 0, None)
    pred_t = np.clip(pred_t, 0, None)

    # Clean up Keras objects
    for _, m in trained_models.items():
        del m
    tf.keras.backend.clear_session()
    gc.collect()

    return {
        "unemployment": pred_u,
        "vacancies": pred_v,
        "theta": pred_t
    }

# ============================================================
# 8. MODEL-SPECIFIC FORECAST FUNCTIONS
# ============================================================

def forecast_sarimax(train_df, horizon):
    # Unemployment
    u_model = SARIMAX(
        train_df["log_unemployment"],
        order=SARIMAX_U_ORDER,
        seasonal_order=SARIMAX_U_SEASONAL,
        trend="n",
        enforce_stationarity=False,
        enforce_invertibility=False
    )
    u_fit = u_model.fit(disp=False)
    u_fc_log = np.asarray(u_fit.get_forecast(steps=horizon).predicted_mean)
    u_fc = np.expm1(u_fc_log)
    u_fc = np.clip(u_fc, 0, None)

    # Vacancies
    v_model = SARIMAX(
        train_df["log_vacancies"],
        order=SARIMAX_V_ORDER,
        seasonal_order=SARIMAX_V_SEASONAL,
        trend="n",
        enforce_stationarity=False,
        enforce_invertibility=False
    )
    v_fit = v_model.fit(disp=False)
    v_fc_log = np.asarray(v_fit.get_forecast(steps=horizon).predicted_mean)
    v_fc = np.expm1(v_fc_log)
    v_fc = np.clip(v_fc, 0, None)

    t_fc = np.where(u_fc > 0, v_fc / u_fc, np.nan)

    return {
        "unemployment": u_fc,
        "vacancies": v_fc,
        "theta": t_fc
    }

def forecast_tbats_prophet(train_df, horizon, forecast_dates):
    # TBATS on unemployment level
    tbats_estimator = TBATS(seasonal_periods=[12])
    u_model = tbats_estimator.fit(train_df["unemployment"].values)
    u_fc = np.asarray(u_model.forecast(steps=horizon))
    u_fc = np.clip(u_fc, 0, None)

    # Prophet on vacancies level
    prophet_train = train_df.reset_index().iloc[:, [0, train_df.columns.get_loc("vacancies") + 1]].copy()
    prophet_train.columns = ["ds", "y"]

    prophet_model = Prophet(
        yearly_seasonality=True,
        weekly_seasonality=False,
        daily_seasonality=False
    )
    prophet_model.fit(prophet_train)

    future = prophet_model.make_future_dataframe(periods=horizon, freq="MS")
    prophet_forecast = prophet_model.predict(future)

    # Align last h rows to forecast_dates
    v_fc = prophet_forecast.set_index("ds")["yhat"].reindex(forecast_dates)
    if v_fc.isna().any():
        # fallback: take the final horizon rows if index alignment misses
        v_fc = prophet_forecast["yhat"].tail(horizon).values
    else:
        v_fc = v_fc.values

    v_fc = np.asarray(v_fc, dtype=float)
    v_fc = np.clip(v_fc, 0, None)

    t_fc = np.where(u_fc > 0, v_fc / u_fc, np.nan)

    return {
        "unemployment": u_fc,
        "vacancies": v_fc,
        "theta": t_fc
    }

def forecast_var(train_df, horizon):
    var_df = train_df.copy()

    var_df["d_log_unemployment"] = var_df["log_unemployment"].diff()
    var_df["d_log_vacancies"] = var_df["log_vacancies"].diff()

    var_system = var_df[["d_log_unemployment", "d_log_vacancies"]].dropna()

    lag_selector = VAR(var_system)
    lag_selection = lag_selector.select_order(maxlags=VAR_MAX_LAGS)

    selected_lag = getattr(lag_selection, VAR_LAG_CRITERION, None)

    if selected_lag is None or not isinstance(selected_lag, (int, np.integer)) or selected_lag < 1:
        selected_lag = 1
    selected_lag = int(selected_lag)

    var_model = VAR(var_system)
    var_fit = var_model.fit(selected_lag)

    fc_input = var_system.values[-selected_lag:]
    fc_diff = var_fit.forecast(fc_input, steps=horizon)

    fc_diff = pd.DataFrame(fc_diff, columns=["d_log_unemployment", "d_log_vacancies"])

    last_log_u = train_df["log_unemployment"].iloc[-1]
    last_log_v = train_df["log_vacancies"].iloc[-1]

    fc_log_u = last_log_u + fc_diff["d_log_unemployment"].cumsum().values
    fc_log_v = last_log_v + fc_diff["d_log_vacancies"].cumsum().values

    fc_u = np.expm1(fc_log_u)
    fc_v = np.expm1(fc_log_v)

    fc_u = np.clip(fc_u, 0, None)
    fc_v = np.clip(fc_v, 0, None)

    fc_t = np.where(fc_u > 0, fc_v / fc_u, np.nan)

    return {
        "unemployment": fc_u,
        "vacancies": fc_v,
        "theta": fc_t
    }

# ============================================================
# 9. MAIN ROLLING-ORIGIN LOOP
# ============================================================

prediction_rows = []

for n_origin, origin_idx in enumerate(origin_indices, start=1):
    origin_date = all_dates[origin_idx]

    print(f"\n==================================================")
    print(f"Origin {n_origin}/{len(origin_indices)} | Train end = {origin_date}")
    print("==================================================")

    # Training data includes everything up to the origin date
    train_df = df.loc[:origin_date].copy()

    # Actual future dates for this origin
    future_dates = all_dates[origin_idx + 1 : origin_idx + 1 + MAX_HORIZON]

    # Actuals by target
    actual_future = df.loc[future_dates, ["unemployment", "vacancies", "theta"]].copy()

    # Forecast containers
    model_forecasts = {}

    # -------------------------------
    # SARIMAX
    # -------------------------------
    if RUN_SARIMAX:
        print("Running SARIMAX...")
        try:
            model_forecasts["sarimax"] = forecast_sarimax(train_df, MAX_HORIZON)
        except Exception as e:
            print(f"SARIMAX failed at origin {origin_date}: {e}")
            model_forecasts["sarimax"] = {
                "unemployment": np.repeat(np.nan, MAX_HORIZON),
                "vacancies": np.repeat(np.nan, MAX_HORIZON),
                "theta": np.repeat(np.nan, MAX_HORIZON)
            }

    # -------------------------------
    # TBATS-Prophet
    # -------------------------------
    if RUN_TBATS_PROPHET:
        print("Running TBATS-Prophet...")
        try:
            model_forecasts["tbats_prophet"] = forecast_tbats_prophet(train_df, MAX_HORIZON, future_dates)
        except Exception as e:
            print(f"TBATS-Prophet failed at origin {origin_date}: {e}")
            model_forecasts["tbats_prophet"] = {
                "unemployment": np.repeat(np.nan, MAX_HORIZON),
                "vacancies": np.repeat(np.nan, MAX_HORIZON),
                "theta": np.repeat(np.nan, MAX_HORIZON)
            }

    # -------------------------------
    # VAR
    # -------------------------------
    if RUN_VAR:
        print("Running VAR...")
        try:
            model_forecasts["var"] = forecast_var(train_df, MAX_HORIZON)
        except Exception as e:
            print(f"VAR failed at origin {origin_date}: {e}")
            model_forecasts["var"] = {
                "unemployment": np.repeat(np.nan, MAX_HORIZON),
                "vacancies": np.repeat(np.nan, MAX_HORIZON),
                "theta": np.repeat(np.nan, MAX_HORIZON)
            }

    # -------------------------------
    # BiLSTM
    # -------------------------------
    if RUN_BILSTM:
        print("Running BiLSTM-Attention...")
        try:
            model_forecasts["bilstm"] = forecast_bilstm_recursive(train_df, MAX_HORIZON)
        except Exception as e:
            print(f"BiLSTM failed at origin {origin_date}: {e}")
            model_forecasts["bilstm"] = {
                "unemployment": np.repeat(np.nan, MAX_HORIZON),
                "vacancies": np.repeat(np.nan, MAX_HORIZON),
                "theta": np.repeat(np.nan, MAX_HORIZON)
            }

    # -------------------------------
    # Store rows for each target/horizon
    # -------------------------------
    for h in range(1, MAX_HORIZON + 1):
        forecast_date = future_dates[h - 1]

        for target in ["unemployment", "vacancies", "theta"]:
            row = {
                "target": target,
                "origin": origin_date,
                "forecast_date": forecast_date,
                "horizon": h,
                "actual": actual_future.loc[forecast_date, target],
                "sarimax_pred": model_forecasts["sarimax"][target][h - 1] if RUN_SARIMAX else np.nan,
                "tbats_prophet_pred": model_forecasts["tbats_prophet"][target][h - 1] if RUN_TBATS_PROPHET else np.nan,
                "var_pred": model_forecasts["var"][target][h - 1] if RUN_VAR else np.nan,
                "bilstm_pred": model_forecasts["bilstm"][target][h - 1] if RUN_BILSTM else np.nan,
                "train_end_date": origin_date
            }
            prediction_rows.append(row)

    gc.collect()

# ============================================================
# 10. MASTER ROLLING PREDICTIONS TABLE
# ============================================================

rolling_predictions = pd.DataFrame(prediction_rows)

rolling_predictions["origin"] = pd.to_datetime(rolling_predictions["origin"])
rolling_predictions["forecast_date"] = pd.to_datetime(rolling_predictions["forecast_date"])
rolling_predictions["train_end_date"] = pd.to_datetime(rolling_predictions["train_end_date"])

print("\nMaster rolling predictions preview:")
display(rolling_predictions.head(12))

# ============================================================
# 11. SPLIT BY TARGET AND EXPORT MASTER FILES
# ============================================================

target_files = {}
for target in ["unemployment", "vacancies", "theta"]:
    target_df = rolling_predictions[rolling_predictions["target"] == target].copy()

    out_path = EXPORT_DIR / f"rolling_predictions_{target}.csv"
    target_df.to_csv(out_path, index=False)
    target_files[target] = target_df

    print(f"Saved: {out_path}")

# ============================================================
# 12. COMPARISON SUMMARY TABLES
# ============================================================

def build_model_comparison(df_target):
    model_map = {
        "SARIMAX": "sarimax_pred",
        "TBATS_PROPHET": "tbats_prophet_pred",
        "VAR": "var_pred",
        "BiLSTM_Attention": "bilstm_pred"
    }

    rows = []

    for model_name, pred_col in model_map.items():
        y_true = df_target["actual"].values
        y_pred = df_target[pred_col].values

        rows.append({
            "model": model_name,
            "RMSE": safe_rmse(y_true, y_pred),
            "MAE": safe_mae(y_true, y_pred),
            "MAPE": safe_mape(y_true, y_pred),
            "sMAPE": safe_smape(y_true, y_pred),
            "MASE": safe_mase(y_true, y_pred),
            "n_obs": len(clean_metric_inputs(y_true, y_pred)[0])
        })

    summary = pd.DataFrame(rows)

    # Average rank across the 5 metrics
    for metric in ["RMSE", "MAE", "MAPE", "sMAPE", "MASE"]:
        summary[f"rank_{metric}"] = summary[metric].rank(method="min")

    summary["average_rank"] = summary[
        [f"rank_{m}" for m in ["RMSE", "MAE", "MAPE", "sMAPE", "MASE"]]
    ].mean(axis=1)

    summary = summary.sort_values("average_rank").reset_index(drop=True)
    return summary

comparison_tables = {}

for target in ["unemployment", "vacancies", "theta"]:
    summary_df = build_model_comparison(target_files[target])
    comparison_tables[target] = summary_df

    out_path = EXPORT_DIR / f"model_comparison_{target}.csv"
    summary_df.to_csv(out_path, index=False)

    print(f"\nModel comparison — {target}")
    display(summary_df)

# ============================================================
# 13. HORIZON-WISE SUMMARY TABLES
# ============================================================

def build_horizonwise_summary(df_target):
    model_map = {
        "SARIMAX": "sarimax_pred",
        "TBATS_PROPHET": "tbats_prophet_pred",
        "VAR": "var_pred",
        "BiLSTM_Attention": "bilstm_pred"
    }

    rows = []

    for h in sorted(df_target["horizon"].unique()):
        temp = df_target[df_target["horizon"] == h].copy()

        for model_name, pred_col in model_map.items():
            y_true = temp["actual"].values
            y_pred = temp[pred_col].values

            rows.append({
                "horizon": h,
                "model": model_name,
                "RMSE": safe_rmse(y_true, y_pred),
                "MAE": safe_mae(y_true, y_pred),
                "MAPE": safe_mape(y_true, y_pred),
                "sMAPE": safe_smape(y_true, y_pred),
                "MASE": safe_mase(y_true, y_pred),
                "n_obs": len(clean_metric_inputs(y_true, y_pred)[0])
            })

    return pd.DataFrame(rows)

horizonwise_tables = {}
for target in ["unemployment", "vacancies", "theta"]:
    h_df = build_horizonwise_summary(target_files[target])
    horizonwise_tables[target] = h_df

    out_path = EXPORT_DIR / f"horizonwise_comparison_{target}.csv"
    h_df.to_csv(out_path, index=False)

# ============================================================
# 14. TABLE PREVIEWS
# ============================================================

print("\nUnemployment rolling predictions preview:")
display(target_files["unemployment"].head(10))

print("\nVacancies rolling predictions preview:")
display(target_files["vacancies"].head(10))

print("\nTheta rolling predictions preview:")
display(target_files["theta"].head(10))

# ============================================================
# 15. VISUALS — HORIZON 1 ACTUAL VS PREDICTIONS
# ============================================================

for target in ["unemployment", "vacancies", "theta"]:
    plot_df = target_files[target].copy()
    plot_df = plot_df[plot_df["horizon"] == 1].sort_values("forecast_date")

    plt.figure(figsize=(14, 6))

    plt.plot(plot_df["forecast_date"], plot_df["actual"], label="Actual", linewidth=2)
    plt.plot(plot_df["forecast_date"], plot_df["sarimax_pred"], label="SARIMAX", linestyle="--")
    plt.plot(plot_df["forecast_date"], plot_df["tbats_prophet_pred"], label="TBATS-Prophet", linestyle="--")
    plt.plot(plot_df["forecast_date"], plot_df["var_pred"], label="VAR", linestyle="--")
    plt.plot(plot_df["forecast_date"], plot_df["bilstm_pred"], label="BiLSTM", linestyle="--")

    plt.title(f"Rolling-Origin Backtesting — Horizon 1 — {target}")
    plt.xlabel("Forecast Date")
    plt.ylabel(target)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# 16. VISUALS — RMSE BAR CHART BY TARGET
# ============================================================

for target in ["unemployment", "vacancies", "theta"]:
    summary_df = comparison_tables[target].copy()

    plt.figure(figsize=(10, 5))
    plt.bar(summary_df["model"], summary_df["RMSE"])
    plt.title(f"RMSE Comparison — {target}")
    plt.xlabel("Model")
    plt.ylabel("RMSE")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()
    plt.show()

# ============================================================
# 17. FINAL STATUS MESSAGE
# ============================================================

print("\n============================================================")
print("ROLLING-ORIGIN BACKTESTING COMPLETED")
print("============================================================")
print("Files saved to:", EXPORT_DIR.resolve())
print("\nSaved master files:")
print(" - rolling_predictions_unemployment.csv")
print(" - rolling_predictions_vacancies.csv")
print(" - rolling_predictions_theta.csv")
print(" - model_comparison_unemployment.csv")
print(" - model_comparison_vacancies.csv")
print(" - model_comparison_theta.csv")
print(" - horizonwise_comparison_unemployment.csv")
print(" - horizonwise_comparison_vacancies.csv")
print(" - horizonwise_comparison_theta.csv")
```
