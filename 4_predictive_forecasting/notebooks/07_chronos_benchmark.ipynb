```python
# =============================================================================
# CHRONOS BENCHMARK — TRAIN / TEST EVALUATION
# =============================================================================
#
# Purpose
# -------
# Produce a paper-consistent Chronos benchmark over a single train → test
# split for three labour-market targets:
#
#   • unemployment
#   • vacancies
#   • theta   (vacancy-unemployment ratio, V/U)
#
# Two benchmark variants are built:
#
#   A) Chronos Univariate
#      One independent ChronosPipeline forecast per target.  This is the
#      canonical, paper-consistent benchmark.
#
#   B) Chronos Serialised Multivariate  (experimental)
#      All three series are z-scored and interleaved into a single 1-D
#      sequence  [u_1, v_1, θ_1, u_2, v_2, θ_2, …].  Chronos forecasts
#      3 × H steps and the block is reshaped back to H × 3 then
#      inverse-standardised.  This captures cross-series co-movement
#      within the zero-shot Chronos framework.
#
# Outputs
# -------
#   exports/chronos/   — forecast CSVs + metric summary
#   figures/chronos/   — train/test split plots + test-period comparison plots
#
# Data Loading
# ------------
# Uses the project-canonical data_loader.py at:
#   <project_root>/4_predictive_forecasting/utils/data_loader.py
#
# Project root is resolved as two levels above the current working directory:
#   Path.cwd().parents[1]
#
# =============================================================================


# =============================================================================
# SECTION 1 — IMPORTS & ENVIRONMENT SETUP
# =============================================================================

import warnings
warnings.filterwarnings("ignore")

import gc
import os
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.dates as mdates

from IPython.display import display

# ── PyTorch ──────────────────────────────────────────────────────────────────
try:
    import torch
except ImportError as exc:
    raise ImportError(
        "PyTorch is required.  Run:  pip install torch"
    ) from exc

# ── Chronos ──────────────────────────────────────────────────────────────────
# The modern chronos-forecasting package (v1.4+) exposes BaseChronosPipeline
# as the unified entry point for all model families (Chronos, Chronos-Bolt,
# Chronos-2).  ChronosPipeline is kept for backwards compatibility but its
# predict() signature changed — context is now positional, not a keyword arg.
# We import both so the version-detection logic below can pick the right path.
try:
    from chronos import BaseChronosPipeline          # v1.4+  (preferred)
    _CHRONOS_BASE_AVAILABLE = True
except ImportError:
    _CHRONOS_BASE_AVAILABLE = False

try:
    from chronos import ChronosPipeline              # legacy fallback
    _CHRONOS_LEGACY_AVAILABLE = True
except ImportError:
    _CHRONOS_LEGACY_AVAILABLE = False

if not _CHRONOS_BASE_AVAILABLE and not _CHRONOS_LEGACY_AVAILABLE:
    raise ImportError(
        "Chronos is not installed.  Run:\n"
        "  pip install chronos-forecasting\n"
        "or for the latest version:\n"
        "  pip install git+https://github.com/amazon-science/chronos-forecasting.git"
    )

# ── Display options ──────────────────────────────────────────────────────────
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 220)
pd.set_option("display.float_format", "{:.4f}".format)

# ── CPU thread cap (notebook stability) ──────────────────────────────────────
try:
    _cpu_threads = max(1, min(4, (os.cpu_count() or 2) // 2))
    torch.set_num_threads(_cpu_threads)
except Exception:
    pass


# =============================================================================
# SECTION 2 — USER-CONFIGURABLE PARAMETERS
# =============================================================================

# Chronos model size.  Options: tiny | mini | small | base | large
#   'small' balances accuracy and memory for most machines.
MODEL_NAME: str = "amazon/chronos-t5-small"

# Number of Monte-Carlo sample paths drawn from the Chronos predictive
# distribution.  Median of these paths is used as the point forecast.
NUM_SAMPLES: int = 20

# Number of months reserved as the held-out test set when no pre-split
# train/test variables exist in memory.
TEST_HORIZON: int = 12

# Optional hard split date (ISO string).  If set, overrides TEST_HORIZON.
# Example: FIXED_TEST_START = "2024-01-01"
FIXED_TEST_START = None

# Target series names — must match column names in the loaded dataset.
TARGETS: list[str] = ["unemployment", "vacancies", "theta"]

# Output directories
BASE_DIR   = Path(".")
EXPORT_DIR = BASE_DIR / "exports" / "chronos"
FIG_DIR    = BASE_DIR / "figures" / "chronos"

EXPORT_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

# ── Device selection ──────────────────────────────────────────────────────────
# bfloat16 is used for all Chronos model loads — numerically stable on both
# CPU and CUDA, and matches the official Chronos documentation examples.
DEVICE_MAP = "cuda" if torch.cuda.is_available() else "cpu"


# =============================================================================
# SECTION 3 — LOAD CHRONOS PIPELINE
# =============================================================================

print("=" * 95)
print("STEP 1 — LOADING CHRONOS PIPELINE")
print("=" * 95)
print(f"  Model  : {MODEL_NAME}")
print(f"  Device : {DEVICE_MAP}")
print(f"  Dtype  : {TORCH_DTYPE}")
print()

# BaseChronosPipeline.from_pretrained auto-detects model family (Chronos /
# Chronos-Bolt / Chronos-2) and returns the correct pipeline subclass.
# bfloat16 is strongly preferred over float16 — it avoids overflow on CPU
# and is numerically stable across all Chronos model sizes.
_load_dtype = torch.bfloat16   # stable on both CPU and CUDA for Chronos

if _CHRONOS_BASE_AVAILABLE:
    pipeline = BaseChronosPipeline.from_pretrained(
        MODEL_NAME,
        device_map=DEVICE_MAP,
        torch_dtype=_load_dtype,
    )
    _pipeline_variant = "BaseChronosPipeline (v1.4+)"
else:
    # Legacy path — ChronosPipeline still works on older package versions
    pipeline = ChronosPipeline.from_pretrained(
        MODEL_NAME,
        device_map=DEVICE_MAP,
        torch_dtype=_load_dtype,
    )
    _pipeline_variant = "ChronosPipeline (legacy)"

print(f"  Pipeline variant : {_pipeline_variant}")
print("  ✓ Chronos pipeline loaded.\n")


# =============================================================================
# SECTION 4 — DATASET LOADING VIA PROJECT DATA LOADER
# =============================================================================

print("=" * 95)
print("STEP 2 — LOADING DATASET")
print("=" * 95)

# ── Resolve project root and append utils to sys.path ────────────────────────
PROJECT_ROOT = Path.cwd().parents[1]
UTILS_PATH   = PROJECT_ROOT / "4_predictive_forecasting" / "utils"

if str(UTILS_PATH) not in sys.path:
    sys.path.append(str(UTILS_PATH))

print(f"  Project root : {PROJECT_ROOT}")
print(f"  Utils path   : {UTILS_PATH}\n")

# ── Import and call the canonical data loader ─────────────────────────────────
from data_loader import load_dataset          # noqa: E402  (appended to path above)

df_full = load_dataset(verbose=True)

# Ensure the index is a proper DatetimeIndex named 'month'
if not isinstance(df_full.index, pd.DatetimeIndex):
    raise TypeError(
        "load_dataset() must return a DataFrame with a DatetimeIndex.  "
        "Check data_loader.py."
    )

df_full = df_full.sort_index()
df_full.index.name = "month"

# ── Validate required target columns ─────────────────────────────────────────
_missing = [c for c in TARGETS if c not in df_full.columns]
if _missing:
    raise ValueError(
        f"Dataset is missing required target columns: {_missing}\n"
        f"Available columns: {list(df_full.columns)}"
    )

# Convert targets to float and drop rows where any target is NaN
for _col in TARGETS:
    df_full[_col] = pd.to_numeric(df_full[_col], errors="coerce")

df_full = df_full.dropna(subset=TARGETS).copy()

print(f"\n  ✓ Full dataset shape after cleaning : {df_full.shape}")
print(f"  ✓ Period : {df_full.index.min().date()} → {df_full.index.max().date()}\n")


# =============================================================================
# SECTION 5 — TRAIN / TEST SPLIT
# =============================================================================

print("=" * 95)
print("STEP 3 — TRAIN / TEST SPLIT")
print("=" * 95)

if FIXED_TEST_START is not None:
    # Hard calendar split
    _split = pd.Timestamp(FIXED_TEST_START)
    train_df = df_full.loc[df_full.index < _split].copy()
    test_df  = df_full.loc[df_full.index >= _split].copy()
    print(f"  Split mode : fixed date  ({FIXED_TEST_START})")
else:
    # Last-N split
    if len(df_full) <= TEST_HORIZON:
        raise ValueError(
            f"Dataset has only {len(df_full)} rows; "
            f"cannot reserve {TEST_HORIZON} as test set."
        )
    train_df = df_full.iloc[:-TEST_HORIZON].copy()
    test_df  = df_full.iloc[-TEST_HORIZON:].copy()
    print(f"  Split mode : last {TEST_HORIZON} months")

# Guard: test must start strictly after train ends
if train_df.index.max() >= test_df.index.min():
    test_df = test_df.loc[test_df.index > train_df.index.max()].copy()

if len(train_df) < 12:
    raise ValueError(f"Training set too short ({len(train_df)} rows).  Need ≥ 12.")
if len(test_df) == 0:
    raise ValueError("Test set is empty after split.  Adjust TEST_HORIZON or FIXED_TEST_START.")

FORECAST_H = len(test_df)   # prediction length equals test-set length

print(f"  Train rows    : {len(train_df):,}")
print(f"  Test rows     : {len(test_df):,}  (H = {FORECAST_H})")
print(f"  Train period  : {train_df.index.min().date()} → {train_df.index.max().date()}")
print(f"  Test period   : {test_df.index.min().date()} → {test_df.index.max().date()}")

print("\n  Train sample:")
display(train_df[TARGETS].tail(3))
print("  Test sample:")
display(test_df[TARGETS].head(3))


# =============================================================================
# SECTION 6 — HELPER FUNCTIONS
# =============================================================================

# ── 6A. Chronos inference wrapper ────────────────────────────────────────────

def chronos_point_forecast(
    series: np.ndarray,
    horizon: int,
    num_samples: int = NUM_SAMPLES,
) -> np.ndarray:
    """
    Run Chronos on a 1-D numpy array and return a point forecast of length
    `horizon`.

    API compatibility notes
    -----------------------
    All known versions of chronos-forecasting (v1.x) accept:
        predict(context, prediction_length, num_samples)
    where ALL three arguments must be passed POSITIONALLY.

    Passing any of them as keyword arguments causes TypeError because the
    internal parameter name differs across versions:
        • v1.0–v1.3  →  'context'
        • v1.4+      →  'inputs'
        • predict_quantiles also uses 'inputs', not 'context'

    The safe, version-agnostic call is always fully positional.
    Median across sample paths is used as the point forecast.

    Parameters
    ----------
    series      : 1-D float array — the conditioning context.
    horizon     : Number of future steps to forecast.
    num_samples : Monte-Carlo samples drawn from the predictive distribution.

    Returns
    -------
    forecast : 1-D numpy array of length `horizon`.
    """
    if len(series) < 2:
        raise ValueError("Context series must have at least 2 observations.")

    # Build a 1-D float32 tensor — required by all Chronos pipeline variants
    context_tensor = torch.tensor(series.astype(np.float32))

    # ── All Chronos versions: use predict() with fully positional args ─────────
    # Both ChronosPipeline and BaseChronosPipeline accept:
    #   predict(context, prediction_length, num_samples)
    # where context is a tensor OR a list of tensors.
    # NEVER pass these as keyword arguments — the internal signature uses
    # 'inputs' not 'context' in some versions, causing TypeError either way.
    raw = pipeline.predict(
        context_tensor,          # positional arg 1 — the context tensor
        int(horizon),            # positional arg 2 — prediction_length
        int(num_samples),        # positional arg 3 — num_samples
    )
    arr = raw.detach().cpu().numpy() if torch.is_tensor(raw) else np.asarray(raw)

    # Normalise to (num_samples, horizon) then take median
    if arr.ndim == 3:            # (batch=1, num_samples, horizon)
        arr = arr[0]
    if arr.ndim == 2:            # (num_samples, horizon)  ← standard
        arr = np.median(arr, axis=0)
    # arr.ndim == 1 → already (horizon,)

    forecast = np.asarray(arr, dtype=float).reshape(-1)

    if len(forecast) < horizon:
        raise ValueError(
            f"Chronos returned {len(forecast)} steps but {horizon} were requested."
        )

    return forecast[:horizon]


# ── 6B. Standardisation helpers ───────────────────────────────────────────────

def z_score(x: np.ndarray) -> tuple[np.ndarray, float, float]:
    """Z-score a 1-D array.  Returns (z, mean, std)."""
    mu = float(np.nanmean(x))
    sd = float(np.nanstd(x))
    sd = sd if (np.isfinite(sd) and not np.isclose(sd, 0.0)) else 1.0
    return (x - mu) / sd, mu, sd


def inv_z_score(z: np.ndarray, mu: float, sd: float) -> np.ndarray:
    """Inverse z-score."""
    return np.asarray(z, dtype=float) * sd + mu


# ── 6C. Forecast-error metrics ────────────────────────────────────────────────

def _rmse(y: np.ndarray, yhat: np.ndarray) -> float:
    return float(np.sqrt(np.mean((y - yhat) ** 2)))

def _mae(y: np.ndarray, yhat: np.ndarray) -> float:
    return float(np.mean(np.abs(y - yhat)))

def _mape(y: np.ndarray, yhat: np.ndarray) -> float:
    mask = np.abs(y) > 1e-12
    if mask.sum() == 0:
        return np.nan
    return float(np.mean(np.abs((y[mask] - yhat[mask]) / y[mask])) * 100)

def _smape(y: np.ndarray, yhat: np.ndarray) -> float:
    denom = np.abs(y) + np.abs(yhat)
    mask  = denom > 1e-12
    if mask.sum() == 0:
        return np.nan
    return float(np.mean(2 * np.abs(yhat[mask] - y[mask]) / denom[mask]) * 100)

def _mase(y: np.ndarray, yhat: np.ndarray, train_series: np.ndarray) -> float:
    """
    MASE scale is the mean absolute naive (random-walk) error on the training
    series.  A value < 1 means the model beats the naïve benchmark.
    """
    diffs = np.abs(np.diff(train_series))
    diffs = diffs[np.isfinite(diffs)]
    if len(diffs) == 0 or np.isclose(diffs.mean(), 0.0):
        return np.nan
    return float(_mae(y, yhat) / diffs.mean())

def compute_metrics(
    actual: np.ndarray,
    predicted: np.ndarray,
    train_series: np.ndarray,
    model_name: str,
    target_name: str,
) -> dict:
    """Return a metrics dictionary for one forecast."""
    y    = np.asarray(actual,    dtype=float)
    yhat = np.asarray(predicted, dtype=float)
    return {
        "target"     : target_name,
        "model"      : model_name,
        "RMSE"       : _rmse(y, yhat),
        "MAE"        : _mae(y, yhat),
        "MAPE (%)"   : _mape(y, yhat),
        "sMAPE (%)"  : _smape(y, yhat),
        "MASE"       : _mase(y, yhat, train_series),
        "n_forecasts": int(np.isfinite(y).sum()),
    }


# ── 6D. Plotting helpers ───────────────────────────────────────────────────────

_TARGET_LABELS = {
    "unemployment": "Unemployment (thousands)",
    "vacancies"   : "Vacancies (thousands)",
    "theta"       : "θ  =  V / U  (vacancy–unemployment ratio)",
}

def _fmt_xaxis(ax):
    ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))
    ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
    plt.setp(ax.xaxis.get_majorticklabels(), rotation=45, ha="right")


def plot_train_test_forecast(
    train_df: pd.DataFrame,
    test_df:  pd.DataFrame,
    pred_uni: np.ndarray,
    pred_mv:  np.ndarray,
    target:   str,
    save_dir: Path,
):
    """
    Full-series view:  training history + test actuals + both Chronos forecasts.
    """
    fig, ax = plt.subplots(figsize=(15, 5))

    ax.plot(train_df.index, train_df[target],
            color="steelblue", linewidth=2, label="Train (actual)")
    ax.plot(test_df.index, test_df[target],
            color="black", linewidth=2, label="Test (actual)")
    ax.plot(test_df.index, pred_uni,
            color="crimson",  linewidth=2, linestyle="--",
            label="Chronos Univariate")
    ax.plot(test_df.index, pred_mv,
            color="darkorange", linewidth=2, linestyle="-.",
            label="Chronos Serialised Multivariate")

    # Shaded test region
    ax.axvspan(test_df.index.min(), test_df.index.max(),
               alpha=0.06, color="grey", label="Test window")

    ax.set_title(
        f"{target.title()} — Train / Test Split with Chronos Forecasts",
        fontsize=13, fontweight="bold"
    )
    ax.set_xlabel("Month", fontsize=11)
    ax.set_ylabel(_TARGET_LABELS.get(target, target), fontsize=11)
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)
    _fmt_xaxis(ax)

    fig.tight_layout()
    out_path = save_dir / f"{target}_train_test_forecast.png"
    fig.savefig(out_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"  Saved: {out_path}")


def plot_test_comparison(
    test_df:  pd.DataFrame,
    pred_uni: np.ndarray,
    pred_mv:  np.ndarray,
    target:   str,
    save_dir: Path,
):
    """
    Zoomed test-period view: actuals vs both forecasts + error bands.
    """
    fig, axes = plt.subplots(2, 1, figsize=(14, 9),
                             gridspec_kw={"height_ratios": [3, 1]})

    # ── Top panel: levels ────────────────────────────────────────────────────
    ax = axes[0]
    ax.plot(test_df.index, test_df[target],
            color="black", linewidth=2.5, marker="o", markersize=4,
            label="Actual")
    ax.plot(test_df.index, pred_uni,
            color="crimson", linewidth=2, linestyle="--", marker="s", markersize=3,
            label="Chronos Univariate")
    ax.plot(test_df.index, pred_mv,
            color="darkorange", linewidth=2, linestyle="-.", marker="^", markersize=3,
            label="Chronos Serialised Multivariate")

    ax.set_title(
        f"{target.title()} — Test-Period Forecast Comparison",
        fontsize=13, fontweight="bold"
    )
    ax.set_ylabel(_TARGET_LABELS.get(target, target), fontsize=11)
    ax.legend(fontsize=9)
    ax.grid(True, alpha=0.3)
    _fmt_xaxis(ax)

    # ── Bottom panel: forecast errors ────────────────────────────────────────
    err_uni = test_df[target].values - pred_uni
    err_mv  = test_df[target].values - pred_mv

    ax2 = axes[1]
    ax2.bar(test_df.index, err_uni,
            width=20, color="crimson", alpha=0.5, label="Error — Univariate")
    ax2.bar(test_df.index, err_mv,
            width=20, color="darkorange", alpha=0.5, label="Error — Multivariate")
    ax2.axhline(0, color="black", linewidth=0.8)
    ax2.set_ylabel("Forecast Error", fontsize=10)
    ax2.set_xlabel("Month", fontsize=11)
    ax2.legend(fontsize=8)
    ax2.grid(True, alpha=0.3)
    _fmt_xaxis(ax2)

    fig.tight_layout()
    out_path = save_dir / f"{target}_test_forecast_comparison.png"
    fig.savefig(out_path, dpi=300, bbox_inches="tight")
    plt.show()
    print(f"  Saved: {out_path}")


# =============================================================================
# SECTION 7 — CHRONOS UNIVARIATE FORECASTS
# =============================================================================

print("\n" + "=" * 95)
print("STEP 4 — CHRONOS UNIVARIATE FORECASTS")
print("=" * 95)
print(
    "  Each target series is forecast independently.\n"
    "  The full training set is used as the conditioning context.\n"
    "  Point forecast = median over all Monte-Carlo sample paths.\n"
)

uni_preds: dict[str, np.ndarray] = {}

for target in TARGETS:
    print(f"  Forecasting  [{target}]  …", end="  ")

    context = train_df[target].values.astype(float)

    preds = chronos_point_forecast(
        series=context,
        horizon=FORECAST_H,
        num_samples=NUM_SAMPLES,
    )

    uni_preds[target] = preds

    _preview = pd.DataFrame({
        "month"            : test_df.index,
        "actual"           : test_df[target].values,
        "chronos_uni_pred" : preds,
        "error"            : test_df[target].values - preds,
    })
    print(f"done  (RMSE = {_rmse(test_df[target].values, preds):.4f})")
    display(_preview.head())

# Free GPU / CPU memory between benchmark steps
if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()

print("\n  ✓ Univariate forecasts complete.\n")


# =============================================================================
# SECTION 8 — CHRONOS SERIALISED MULTIVARIATE FORECASTS
# =============================================================================
#
# Methodology
# -----------
# 1. Z-score each training series independently (prevents scale dominance).
# 2. Interleave into one long 1-D context vector:
#      [u₁, v₁, θ₁, u₂, v₂, θ₂, …, u_T, v_T, θ_T]
# 3. Request Chronos to forecast 3 × H future steps.
# 4. Reshape the forecast block from (3H,) back to (H, 3).
# 5. Inverse z-score each column using the training-set scalers.
#
# This is an experimental approach — the model has no explicit knowledge of
# the three-series structure, but cross-series temporal co-movement is
# implicitly encoded in the interleaved context.
#
# =============================================================================

print("=" * 95)
print("STEP 5 — CHRONOS SERIALISED MULTIVARIATE FORECASTS")
print("=" * 95)
print(
    "  Series order in the interleaved context:\n"
    "    [unemployment, vacancies, theta] repeated for each training month.\n"
)

# ── Step 8A: z-score each training series and collect scalers ─────────────────
z_train: dict[str, np.ndarray] = {}
scalers: dict[str, dict]       = {}

for target in TARGETS:
    z, mu, sd = z_score(train_df[target].values.astype(float))
    z_train[target]  = z
    scalers[target]  = {"mu": mu, "sd": sd}
    print(f"  Scaler [{target}]   μ = {mu:.4f},  σ = {sd:.4f}")

# ── Step 8B: build serialised 1-D context ─────────────────────────────────────
# Shape: (T, 3) → flatten row-major → (3T,)
z_matrix    = np.column_stack([z_train[t] for t in TARGETS])   # (T, 3)
serialised  = z_matrix.reshape(-1)                              # (3T,)

print(f"\n  Training observations per series : {len(train_df)}")
print(f"  Serialised context length        : {len(serialised)}  (= 3 × {len(train_df)})")
print(f"  Prediction request length        : {3 * FORECAST_H}  (= 3 × H = 3 × {FORECAST_H})")
print("\n  Running Chronos on serialised context …", end="  ")

# ── Step 8C: single Chronos call ──────────────────────────────────────────────
mv_raw = chronos_point_forecast(
    series=serialised,
    horizon=3 * FORECAST_H,
    num_samples=NUM_SAMPLES,
)

print("done")

# ── Step 8D: reshape and inverse-standardise ──────────────────────────────────
mv_block  = mv_raw[:3 * FORECAST_H].reshape(FORECAST_H, 3)   # (H, 3)

mv_preds: dict[str, np.ndarray] = {}

for i, target in enumerate(TARGETS):
    mv_preds[target] = inv_z_score(
        mv_block[:, i],
        scalers[target]["mu"],
        scalers[target]["sd"],
    )

    _preview = pd.DataFrame({
        "month"           : test_df.index,
        "actual"          : test_df[target].values,
        "chronos_mv_pred" : mv_preds[target],
        "error"           : test_df[target].values - mv_preds[target],
    })
    print(f"\n  Deserialised [{target}]  "
          f"(RMSE = {_rmse(test_df[target].values, mv_preds[target]):.4f})")
    display(_preview.head())

if torch.cuda.is_available():
    torch.cuda.empty_cache()
gc.collect()

print("\n  ✓ Serialised multivariate forecasts complete.\n")


# =============================================================================
# SECTION 9 — MERGE PREDICTIONS INTO BENCHMARK TABLES
# =============================================================================

print("=" * 95)
print("STEP 6 — ASSEMBLING BENCHMARK TABLES")
print("=" * 95)

benchmark: dict[str, pd.DataFrame] = {}

for target in TARGETS:
    df_bm = pd.DataFrame({
        "month"            : test_df.index,
        "actual"           : test_df[target].values.astype(float),
        "chronos_uni_pred" : uni_preds[target].astype(float),
        "chronos_mv_pred"  : mv_preds[target].astype(float),
    })
    df_bm["uni_error"]    = df_bm["actual"] - df_bm["chronos_uni_pred"]
    df_bm["mv_error"]     = df_bm["actual"] - df_bm["chronos_mv_pred"]
    df_bm["uni_abs_error"] = np.abs(df_bm["uni_error"])
    df_bm["mv_abs_error"]  = np.abs(df_bm["mv_error"])
    df_bm["target"]        = target

    benchmark[target] = df_bm.sort_values("month").reset_index(drop=True)

    print(f"\n  Benchmark table — {target}")
    display(benchmark[target].head())

print("\n  ✓ Benchmark tables assembled.\n")


# =============================================================================
# SECTION 10 — FORECAST ACCURACY METRICS
# =============================================================================

print("=" * 95)
print("STEP 7 — FORECAST ACCURACY METRICS")
print("=" * 95)
print(
    "  Metrics computed over the held-out test period.\n"
    "  MASE scale = mean |ΔY| on the training series (naïve random-walk scale).\n"
    "  MASE < 1  →  model beats the naïve random-walk benchmark.\n"
)

metric_rows = []

for target in TARGETS:
    df_bm         = benchmark[target]
    actual        = df_bm["actual"].values
    train_series  = train_df[target].values.astype(float)

    metric_rows.append(
        compute_metrics(actual, df_bm["chronos_uni_pred"].values,
                        train_series, "Chronos Univariate", target)
    )
    metric_rows.append(
        compute_metrics(actual, df_bm["chronos_mv_pred"].values,
                        train_series, "Chronos Serialised Multivariate", target)
    )

chronos_metrics = (
    pd.DataFrame(metric_rows)
    .sort_values(["target", "RMSE"])
    .reset_index(drop=True)
)

print("  Full metric table:\n")
display(chronos_metrics)

# ── Per-target pivot for easy thesis reading ───────────────────────────────────
for target in TARGETS:
    _sub = (
        chronos_metrics.loc[chronos_metrics["target"] == target]
        .drop(columns="target")
        .set_index("model")
    )
    print(f"\n  Metrics — {target.upper()}")
    display(_sub)

print("\n  ✓ Metrics computed.\n")


# =============================================================================
# SECTION 11 — VISUALISATIONS
# =============================================================================

print("=" * 95)
print("STEP 8 — GENERATING THESIS-QUALITY PLOTS")
print("=" * 95)

for target in TARGETS:
    print(f"\n  Plotting [{target}] …")

    # Plot A: full series (train + test + forecasts)
    plot_train_test_forecast(
        train_df=train_df,
        test_df=test_df,
        pred_uni=uni_preds[target],
        pred_mv=mv_preds[target],
        target=target,
        save_dir=FIG_DIR,
    )

    # Plot B: zoomed test-period comparison with error panel
    plot_test_comparison(
        test_df=test_df,
        pred_uni=uni_preds[target],
        pred_mv=mv_preds[target],
        target=target,
        save_dir=FIG_DIR,
    )

print("\n  ✓ All plots saved.\n")


# =============================================================================
# SECTION 12 — EXPORT RESULTS
# =============================================================================

print("=" * 95)
print("STEP 9 — EXPORTING BENCHMARK FILES")
print("=" * 95)

for target in TARGETS:
    df_bm = benchmark[target]

    # Individual variant CSVs
    uni_path  = EXPORT_DIR / f"chronos_univariate_{target}.csv"
    mv_path   = EXPORT_DIR / f"chronos_multivariate_serialised_{target}.csv"
    full_path = EXPORT_DIR / f"chronos_benchmark_merged_{target}.csv"

    df_bm[["month", "actual", "chronos_uni_pred"]].to_csv(uni_path, index=False)
    df_bm[["month", "actual", "chronos_mv_pred"] ].to_csv(mv_path,  index=False)
    df_bm.to_csv(full_path, index=False)

    print(f"  {uni_path}")
    print(f"  {mv_path}")
    print(f"  {full_path}")

# Global metrics CSV
metrics_path = EXPORT_DIR / "chronos_metrics_summary.csv"
chronos_metrics.to_csv(metrics_path, index=False)
print(f"  {metrics_path}")

print("\n  ✓ All exports complete.\n")


# =============================================================================
# SECTION 13 — FINAL AUDIT SUMMARY
# =============================================================================

print("=" * 95)
print("FINAL AUDIT SUMMARY")
print("=" * 95)
print(
    f"  Model        : {MODEL_NAME}\n"
    f"  Pipeline     : {_pipeline_variant}\n"
    f"  Device       : {DEVICE_MAP}\n"
    f"  Samples      : {NUM_SAMPLES}\n"
    f"  Train period : {train_df.index.min().date()} → {train_df.index.max().date()}"
    f"  ({len(train_df)} months)\n"
    f"  Test period  : {test_df.index.min().date()} → {test_df.index.max().date()}"
    f"  (H = {FORECAST_H} months)\n"
)

print(f"  {'Target':<16}  {'Model':<36}  {'RMSE':>10}  {'MASE':>8}  {'sMAPE (%)':>10}")
print("  " + "-" * 86)

for _, row in chronos_metrics.iterrows():
    print(
        f"  {row['target']:<16}  {row['model']:<36}  "
        f"{row['RMSE']:>10.4f}  {row['MASE']:>8.4f}  {row['sMAPE (%)']:>10.4f}"
    )

print("\n  ✓ Chronos benchmark complete.")
print("  Next step: compare against ensemble model in the final comparison notebook.")
```
