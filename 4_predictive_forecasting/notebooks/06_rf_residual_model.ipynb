```python
# =============================================================================
# MODEL 5B — ROLLING-ORIGIN RF RESIDUAL HYBRID (LEAKAGE-SAFE, FINAL VERSION)
# =============================================================================
#
# Purpose
# -------
# Build a Random Forest residual correction layer on top of the dynamic ensemble:
#
#     y_hat_final = y_hat_ensemble + e_hat_RF
#
# for:
#   1) unemployment
#   2) vacancies
#   3) theta
#
# Methodological rule
# -------------------
# The RF model is trained ONLY on earlier forecast origins and predicts residuals
# for the current origin. This keeps the hybrid strictly out-of-sample.
#
# Inputs
# ------
#   exports/ensemble/ensemble_unemployment.csv
#   exports/ensemble/ensemble_vacancies.csv
#   exports/ensemble/ensemble_theta.csv
#
# Expected core columns
# ---------------------
#   origin, forecast_date, horizon, actual,
#   sarimax_pred, tbats_prophet_pred, var_pred, bilstm_pred,
#   ensemble_pred
#
# Optional but recommended columns
# --------------------------------
#   w_sarimax, w_tbats_prophet, w_var, w_bilstm
#
# Outputs
# -------
#   • rolling-origin RF residual predictions
#   • final hybrid forecasts
#   • metric tables
#   • feature-importance plots
#   • export files for final comparison
#
# =============================================================================

# -----------------------------------------------------------------------------
# 1. Imports and setup
# -----------------------------------------------------------------------------
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.ensemble import RandomForestRegressor
from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# -----------------------------------------------------------------------------
# 2. Paths and file configuration
# -----------------------------------------------------------------------------
BASE_DIR = Path(".")
ENSEMBLE_DIR = BASE_DIR / "exports" / "ensemble"
RF_DIR = BASE_DIR / "exports" / "rf_residual_hybrid"
FIG_DIR = BASE_DIR / "figures" / "rf_residual_hybrid"

RF_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

FILES = {
    "unemployment": ENSEMBLE_DIR / "ensemble_unemployment.csv",
    "vacancies": ENSEMBLE_DIR / "ensemble_vacancies.csv",
    "theta": ENSEMBLE_DIR / "ensemble_theta.csv",
}

BASE_MODEL_COLS = [
    "sarimax_pred",
    "tbats_prophet_pred",
    "var_pred",
    "bilstm_pred",
]

WEIGHT_COLS = [
    "w_sarimax",
    "w_tbats_prophet",
    "w_var",
    "w_bilstm",
]

REQUIRED_COLS = [
    "origin",
    "forecast_date",
    "horizon",
    "actual",
    "ensemble_pred",
] + BASE_MODEL_COLS

# -----------------------------------------------------------------------------
# 3. Utility functions
# -----------------------------------------------------------------------------
def rmse(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.sqrt(np.mean((y - yhat) ** 2))

def mae(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.mean(np.abs(y - yhat))

def mape(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    mask = np.abs(y) > 1e-12
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y[mask] - yhat[mask]) / y[mask])) * 100

def smape(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    denom = np.abs(y) + np.abs(yhat)
    mask = denom > 1e-12
    if mask.sum() == 0:
        return np.nan
    return np.mean(2 * np.abs(yhat[mask] - y[mask]) / denom[mask]) * 100

def mase_scale_from_unique_actuals(df):
    """
    MASE denominator from unique forecast-date actuals.
    This avoids repeated actual duplication across rolling origins.
    """
    actual_series = (
        df[["forecast_date", "actual"]]
        .drop_duplicates()
        .sort_values("forecast_date")
        .copy()
    )

    if len(actual_series) < 2:
        return np.nan

    diffs = np.abs(np.diff(actual_series["actual"].values.astype(float)))
    diffs = diffs[~np.isnan(diffs)]

    if len(diffs) == 0:
        return np.nan

    scale = np.mean(diffs)
    if np.isclose(scale, 0):
        return np.nan

    return float(scale)

def mase(y, yhat, scale):
    if scale is None or np.isnan(scale) or np.isclose(scale, 0):
        return np.nan
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.mean(np.abs(y - yhat)) / scale

def validate_and_prepare(df, name):
    """Validate schema, parse dates, coerce numerics, and sort safely."""
    missing = [c for c in REQUIRED_COLS if c not in df.columns]
    if missing:
        raise ValueError(
            f"{name} is missing required columns: {missing}\n"
            f"Available columns: {list(df.columns)}"
        )

    out = df.copy()

    # Parse dates
    out["origin"] = pd.to_datetime(out["origin"], errors="coerce")
    out["forecast_date"] = pd.to_datetime(out["forecast_date"], errors="coerce")

    # Numeric coercion
    numeric_cols = ["horizon", "actual", "ensemble_pred"] + BASE_MODEL_COLS
    for col in numeric_cols:
        out[col] = pd.to_numeric(out[col], errors="coerce")

    # If weight columns are missing, create equal weights as fallback
    for col in WEIGHT_COLS:
        if col not in out.columns:
            out[col] = 0.25

    # Ensure weight columns are numeric
    for col in WEIGHT_COLS:
        out[col] = pd.to_numeric(out[col], errors="coerce").fillna(0.25)

    # Drop unusable rows
    out = out.dropna(subset=["origin", "forecast_date", "horizon", "actual", "ensemble_pred"]).copy()
    out["horizon"] = out["horizon"].astype(int)

    # Sort chronologically
    out = out.sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)

    # Check duplicate key
    key_cols = ["origin", "forecast_date", "horizon"]
    dup_count = out.duplicated(subset=key_cols).sum()
    if dup_count > 0:
        raise ValueError(
            f"{name} has {dup_count} duplicated forecast keys on {key_cols}. "
            "Fix duplicates before RF residual modeling."
        )

    return out

def add_safe_residual_features(df):
    """
    Create leakage-safe features for RF residual modeling.

    Important:
    ----------
    No feature uses future actual information from the current origin.
    Past residual summary features are created using only earlier origins.
    """
    out = df.copy().sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)

    # Base residual target
    out["residual"] = out["actual"] - out["ensemble_pred"]

    # Calendar features known at prediction time
    out["origin_month"] = out["origin"].dt.month
    out["origin_quarter"] = out["origin"].dt.quarter
    out["forecast_month"] = out["forecast_date"].dt.month
    out["forecast_quarter"] = out["forecast_date"].dt.quarter
    out["forecast_year"] = out["forecast_date"].dt.year

    # Base-model disagreement features known at prediction time
    out["pred_mean"] = out[BASE_MODEL_COLS].mean(axis=1)
    out["pred_std"] = out[BASE_MODEL_COLS].std(axis=1)
    out["pred_min"] = out[BASE_MODEL_COLS].min(axis=1)
    out["pred_max"] = out[BASE_MODEL_COLS].max(axis=1)
    out["pred_range"] = out["pred_max"] - out["pred_min"]

    # Pairwise disagreement indicators
    out["sarimax_minus_var"] = out["sarimax_pred"] - out["var_pred"]
    out["tbats_minus_bilstm"] = out["tbats_prophet_pred"] - out["bilstm_pred"]
    out["ensemble_minus_predmean"] = out["ensemble_pred"] - out["pred_mean"]

    # Origin index for regime-like progression
    origin_order = {d: i for i, d in enumerate(sorted(out["origin"].unique()))}
    out["origin_index"] = out["origin"].map(origin_order).astype(int)

    # -------------------------------------------------------------------------
    # Leakage-safe expanding residual history features
    # -------------------------------------------------------------------------
    # Overall past residual mean/std based only on prior rows
    out["past_resid_mean_all"] = out["residual"].shift(1).expanding().mean()
    out["past_resid_std_all"] = out["residual"].shift(1).expanding().std()

    # Horizon-specific past residual mean/std using only earlier rows in same horizon
    out["past_resid_mean_h"] = (
        out.groupby("horizon")["residual"]
        .transform(lambda s: s.shift(1).expanding().mean())
    )
    out["past_resid_std_h"] = (
        out.groupby("horizon")["residual"]
        .transform(lambda s: s.shift(1).expanding().std())
    )

    # Forecast-month-specific past residual mean using only prior rows
    out["past_resid_mean_fmonth"] = (
        out.groupby("forecast_month")["residual"]
        .transform(lambda s: s.shift(1).expanding().mean())
    )

    # Weight concentration / diversification
    out["weight_entropy_proxy"] = (
        np.abs(out["w_sarimax"] - 0.25) +
        np.abs(out["w_tbats_prophet"] - 0.25) +
        np.abs(out["w_var"] - 0.25) +
        np.abs(out["w_bilstm"] - 0.25)
    )

    # Fill early expanding-feature gaps with neutral values
    expanding_fill_cols = [
        "past_resid_mean_all",
        "past_resid_std_all",
        "past_resid_mean_h",
        "past_resid_std_h",
        "past_resid_mean_fmonth",
    ]
    for col in expanding_fill_cols:
        out[col] = out[col].replace([np.inf, -np.inf], np.nan).fillna(0.0)

    # pred_std may be NaN if row somehow has limited valid preds
    out["pred_std"] = out["pred_std"].fillna(0.0)

    return out

def get_feature_columns():
    """Feature set used by the rolling-origin RF model."""
    return [
        "horizon",
        "origin_month",
        "origin_quarter",
        "forecast_month",
        "forecast_quarter",
        "forecast_year",
        "origin_index",

        "ensemble_pred",
        "sarimax_pred",
        "tbats_prophet_pred",
        "var_pred",
        "bilstm_pred",

        "pred_mean",
        "pred_std",
        "pred_range",
        "sarimax_minus_var",
        "tbats_minus_bilstm",
        "ensemble_minus_predmean",

        "w_sarimax",
        "w_tbats_prophet",
        "w_var",
        "w_bilstm",
        "weight_entropy_proxy",

        "past_resid_mean_all",
        "past_resid_std_all",
        "past_resid_mean_h",
        "past_resid_std_h",
        "past_resid_mean_fmonth",
    ]

def build_rolling_rf_hybrid(df, target_name, min_train_origins=4):
    """
    Rolling-origin RF residual prediction.

    For each current origin:
      - train RF only on rows with origin < current_origin
      - predict residual for rows with origin == current_origin
      - if insufficient training history, use zero correction
    """
    out = df.copy().sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)

    feature_cols = get_feature_columns()
    out["rf_residual_pred"] = 0.0
    out["final_hybrid_pred"] = out["ensemble_pred"].copy()
    out["rf_model_used"] = "fallback_zero"

    unique_origins = sorted(out["origin"].unique())
    feature_importance_records = []

    for current_origin in unique_origins:
        train_mask = out["origin"] < current_origin
        test_mask = out["origin"] == current_origin

        train_df = out.loc[train_mask].copy()
        test_df = out.loc[test_mask].copy()

        n_train_origins = train_df["origin"].nunique()

        # Early origins: no reliable residual model yet
        if n_train_origins < min_train_origins or len(train_df) < 20:
            out.loc[test_mask, "rf_residual_pred"] = 0.0
            out.loc[test_mask, "final_hybrid_pred"] = out.loc[test_mask, "ensemble_pred"]
            out.loc[test_mask, "rf_model_used"] = "fallback_zero"
            continue

        X_train = train_df[feature_cols].copy()
        y_train = train_df["residual"].copy()

        X_test = test_df[feature_cols].copy()

        # Fill missing values using train medians only
        train_medians = X_train.median(numeric_only=True)
        X_train = X_train.fillna(train_medians)
        X_test = X_test.fillna(train_medians)

        # RF model
        rf = RandomForestRegressor(
            n_estimators=400,
            max_depth=8,
            min_samples_leaf=3,
            min_samples_split=6,
            max_features="sqrt",
            bootstrap=True,
            random_state=42,
            n_jobs=-1,
        )

        rf.fit(X_train, y_train)
        preds = rf.predict(X_test)

        out.loc[test_mask, "rf_residual_pred"] = preds
        out.loc[test_mask, "final_hybrid_pred"] = out.loc[test_mask, "ensemble_pred"] + preds
        out.loc[test_mask, "rf_model_used"] = "rolling_rf"

        # Save feature importance for this origin
        imp_df = pd.DataFrame({
            "target": target_name,
            "origin": current_origin,
            "feature": feature_cols,
            "importance": rf.feature_importances_,
        })
        feature_importance_records.append(imp_df)

    if feature_importance_records:
        feat_imp = pd.concat(feature_importance_records, axis=0, ignore_index=True)
    else:
        feat_imp = pd.DataFrame(columns=["target", "origin", "feature", "importance"])

    return out, feat_imp

def evaluate_all_models(df, target_name):
    """Evaluate base models, ensemble, and final hybrid."""
    model_map = {
        "SARIMAX": "sarimax_pred",
        "TBATS-Prophet": "tbats_prophet_pred",
        "VAR": "var_pred",
        "BiLSTM-Attention": "bilstm_pred",
        "Dynamic Weighted Ensemble": "ensemble_pred",
        "RF Residual Hybrid": "final_hybrid_pred",
    }

    scale = mase_scale_from_unique_actuals(df)
    results = []

    for model_name, col in model_map.items():
        tmp = df[["actual", col]].dropna().copy()
        if len(tmp) == 0:
            continue

        y = tmp["actual"].values
        yhat = tmp[col].values

        results.append({
            "target": target_name,
            "model": model_name,
            "RMSE": rmse(y, yhat),
            "MAE": mae(y, yhat),
            "MAPE": mape(y, yhat),
            "sMAPE": smape(y, yhat),
            "MASE": mase(y, yhat, scale),
            "n_forecasts": len(tmp),
        })

    return pd.DataFrame(results).sort_values("RMSE").reset_index(drop=True)

def plot_actual_vs_ensemble_vs_hybrid(df, target_name, save_dir):
    """Plot actual, ensemble, and final hybrid forecasts."""
    plot_df = df.sort_values("forecast_date").copy()

    plt.figure(figsize=(14, 6))
    plt.plot(plot_df["forecast_date"], plot_df["actual"], linewidth=2, label="Actual")
    plt.plot(plot_df["forecast_date"], plot_df["ensemble_pred"], linestyle="--", linewidth=2, label="Dynamic Weighted Ensemble")
    plt.plot(plot_df["forecast_date"], plot_df["final_hybrid_pred"], linestyle="-.", linewidth=2, label="RF Residual Hybrid")
    plt.title(f"{target_name.title()} — Actual vs Ensemble vs RF Residual Hybrid")
    plt.xlabel("Forecast Date")
    plt.ylabel("Value")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_actual_ensemble_hybrid.png", dpi=300, bbox_inches="tight")
    plt.show()

def plot_residual_distribution(df, target_name, save_dir):
    """Compare ensemble residuals vs hybrid residuals."""
    tmp = df.copy()
    tmp["ensemble_residual"] = tmp["actual"] - tmp["ensemble_pred"]
    tmp["hybrid_residual"] = tmp["actual"] - tmp["final_hybrid_pred"]

    plt.figure(figsize=(12, 6))
    plt.hist(tmp["ensemble_residual"].dropna(), bins=30, alpha=0.6, label="Ensemble Residual")
    plt.hist(tmp["hybrid_residual"].dropna(), bins=30, alpha=0.6, label="Hybrid Residual")
    plt.title(f"{target_name.title()} — Residual Distribution: Ensemble vs Hybrid")
    plt.xlabel("Residual")
    plt.ylabel("Frequency")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_residual_distribution.png", dpi=300, bbox_inches="tight")
    plt.show()

def plot_feature_importance(feat_imp_df, target_name, save_dir, top_n=12):
    """Plot average RF feature importance across rolling-origin fits."""
    if feat_imp_df.empty:
        print(f"No feature-importance data available for {target_name}.")
        return

    avg_imp = (
        feat_imp_df.groupby("feature", as_index=False)["importance"]
        .mean()
        .sort_values("importance", ascending=False)
        .head(top_n)
    )

    plt.figure(figsize=(10, 6))
    plt.barh(avg_imp["feature"][::-1], avg_imp["importance"][::-1])
    plt.title(f"{target_name.title()} — Average RF Feature Importance")
    plt.xlabel("Average Importance")
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_rf_feature_importance.png", dpi=300, bbox_inches="tight")
    plt.show()

# -----------------------------------------------------------------------------
# 4. Load and validate ensemble files
# -----------------------------------------------------------------------------
print("=" * 90)
print("STEP 1 — LOADING AND VALIDATING ENSEMBLE FILES")
print("=" * 90)

prepared_data = {}

for target, path in FILES.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing required file: {path}")

    raw_df = pd.read_csv(path)
    df = validate_and_prepare(raw_df, target)
    prepared_data[target] = df

    print(f"\n[{target.upper()}]")
    print(f"File: {path}")
    print(f"Rows loaded: {len(raw_df):,}")
    print(f"Rows prepared: {len(df):,}")
    print(f"Unique origins: {df['origin'].nunique():,}")
    print(f"Horizon range: {df['horizon'].min()} to {df['horizon'].max()}")

print("\nEnsemble schema validation completed successfully.")

# -----------------------------------------------------------------------------
# 5. Feature engineering
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 2 — CREATING LEAKAGE-SAFE RF FEATURES")
print("=" * 90)

feature_ready_data = {}

for target, df in prepared_data.items():
    feature_df = add_safe_residual_features(df)
    feature_ready_data[target] = feature_df

    print(f"\n[{target.upper()}] feature table ready")
    print(f"Rows: {len(feature_df):,}")
    preview_cols = [
        "origin", "forecast_date", "horizon", "actual", "ensemble_pred", "residual",
        "pred_mean", "pred_std", "pred_range",
        "past_resid_mean_all", "past_resid_mean_h"
    ]
    display(feature_df[preview_cols].head())

# -----------------------------------------------------------------------------
# 6. Rolling-origin RF residual hybrid
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 3 — ROLLING-ORIGIN RF RESIDUAL MODELING")
print("=" * 90)

hybrid_outputs = {}
feature_importance_outputs = {}

for target, df in feature_ready_data.items():
    hybrid_df, feat_imp_df = build_rolling_rf_hybrid(
        df=df,
        target_name=target,
        min_train_origins=4
    )

    hybrid_outputs[target] = hybrid_df
    feature_importance_outputs[target] = feat_imp_df

    print(f"\n[{target.upper()}] rolling RF hybrid completed")
    print(f"Rows: {len(hybrid_df):,}")
    print("RF usage summary:")
    print(hybrid_df["rf_model_used"].value_counts(dropna=False))

# -----------------------------------------------------------------------------
# 7. Metric evaluation
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 4 — PERFORMANCE COMPARISON")
print("=" * 90)

all_metrics = []

for target, df in hybrid_outputs.items():
    metrics_df = evaluate_all_models(df, target)
    all_metrics.append(metrics_df)

    print(f"\nPerformance comparison — {target}")
    display(metrics_df)

metrics_all = pd.concat(all_metrics, axis=0, ignore_index=True)
print("\nCombined performance summary")
display(metrics_all.sort_values(["target", "RMSE"]).reset_index(drop=True))

# -----------------------------------------------------------------------------
# 8. Diagnostics plots
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 5 — DIAGNOSTIC VISUALS")
print("=" * 90)

for target in ["unemployment", "vacancies", "theta"]:
    print(f"\nGenerating diagnostics for {target} ...")
    plot_actual_vs_ensemble_vs_hybrid(hybrid_outputs[target], target, FIG_DIR)
    plot_residual_distribution(hybrid_outputs[target], target, FIG_DIR)
    plot_feature_importance(feature_importance_outputs[target], target, FIG_DIR, top_n=12)

# -----------------------------------------------------------------------------
# 9. Export hybrid outputs
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 6 — EXPORTING RF HYBRID OUTPUTS")
print("=" * 90)

EXPORT_MAP = {
    "unemployment": RF_DIR / "rf_hybrid_unemployment.csv",
    "vacancies": RF_DIR / "rf_hybrid_vacancies.csv",
    "theta": RF_DIR / "rf_hybrid_theta.csv",
}

for target, out_path in EXPORT_MAP.items():
    export_cols = [
        "origin", "forecast_date", "horizon", "actual",
        "sarimax_pred", "tbats_prophet_pred", "var_pred", "bilstm_pred",
        "w_sarimax", "w_tbats_prophet", "w_var", "w_bilstm",
        "ensemble_pred",
        "residual",
        "rf_residual_pred",
        "final_hybrid_pred",
        "rf_model_used"
    ]
    hybrid_outputs[target][export_cols].to_csv(out_path, index=False)
    print(f"Exported: {out_path}")

# Export feature importance tables
for target, feat_imp_df in feature_importance_outputs.items():
    feat_imp_df.to_csv(RF_DIR / f"rf_feature_importance_{target}.csv", index=False)

# Export combined metrics
metrics_all.to_csv(RF_DIR / "rf_hybrid_metrics_summary.csv", index=False)

# -----------------------------------------------------------------------------
# 10. Final audit summary
# -----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("FINAL AUDIT SUMMARY")
print("=" * 90)

for target in ["unemployment", "vacancies", "theta"]:
    df = hybrid_outputs[target]

    print(f"\n[{target.upper()}]")
    print(f"Rows exported: {len(df):,}")
    print(f"Missing RF residual predictions: {df['rf_residual_pred'].isna().sum():,}")
    print(f"Missing final hybrid predictions: {df['final_hybrid_pred'].isna().sum():,}")

    ens_rmse = rmse(df["actual"], df["ensemble_pred"])
    hyb_rmse = rmse(df["actual"], df["final_hybrid_pred"])

    print(f"Ensemble RMSE: {ens_rmse:.6f}")
    print(f"Hybrid RMSE:   {hyb_rmse:.6f}")
    print(f"RMSE change:   {hyb_rmse - ens_rmse:+.6f}")

print("\nRolling-origin RF residual hybrid pipeline completed successfully.")
print("Next notebook: final Chronos benchmark comparison on the same fold structure.")
```
