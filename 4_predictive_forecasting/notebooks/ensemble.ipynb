```python
# ============================================================================
# MODEL 5A — DYNAMIC WEIGHTED ENSEMBLE (LEAKAGE-SAFE, PHASE-LOCKED VERSION)
# ============================================================================
#
# Purpose
# -------
# Build a PhD-level weighted ensemble for:
#   1) unemployment
#   2) vacancies
#   3) theta
#
# Base models
# -----------
#   • SARIMAX
#   • TBATS-Prophet
#   • VAR
#   • BiLSTM-Attention
#
# Core methodological rule
# ------------------------
# Weights are computed dynamically at each forecast origin using ONLY
# historically available rolling-origin forecast errors from earlier origins.
#
# This avoids temporal leakage.
#
# Main outputs
# ------------
#   • Dynamic ensemble forecasts
#   • Horizon-aware model weights
#   • Performance comparison tables
#   • Diagnostics plots
#   • Export files for RF residual modeling
#
# Expected input files
# --------------------
# exports/backtesting/rolling_predictions_unemployment.csv
# exports/backtesting/rolling_predictions_vacancies.csv
# exports/backtesting/rolling_predictions_theta.csv
#
# Expected minimum schema
# -----------------------
# origin, forecast_date, horizon, actual,
# sarimax_pred, tbats_prophet_pred, var_pred, bilstm_pred
#
# ============================================================================

# ----------------------------------------------------------------------------
# 1. Imports and global setup
# ----------------------------------------------------------------------------
import warnings
warnings.filterwarnings("ignore")

from pathlib import Path
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from IPython.display import display

pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# ----------------------------------------------------------------------------
# 2. Directory configuration
# ----------------------------------------------------------------------------
BASE_DIR = Path(".")
BACKTEST_DIR = BASE_DIR / "exports" / "backtesting"
ENSEMBLE_DIR = BASE_DIR / "exports" / "ensemble"
FIG_DIR = BASE_DIR / "figures" / "ensemble"

ENSEMBLE_DIR.mkdir(parents=True, exist_ok=True)
FIG_DIR.mkdir(parents=True, exist_ok=True)

FILES = {
    "unemployment": BACKTEST_DIR / "rolling_predictions_unemployment.csv",
    "vacancies": BACKTEST_DIR / "rolling_predictions_vacancies.csv",
    "theta": BACKTEST_DIR / "rolling_predictions_theta.csv",
}

MODEL_COLS = [
    "sarimax_pred",
    "tbats_prophet_pred",
    "var_pred",
    "bilstm_pred",
]

MODEL_NAME_MAP = {
    "sarimax_pred": "SARIMAX",
    "tbats_prophet_pred": "TBATS-Prophet",
    "var_pred": "VAR",
    "bilstm_pred": "BiLSTM-Attention",
}

WEIGHT_COL_MAP = {
    "sarimax_pred": "w_sarimax",
    "tbats_prophet_pred": "w_tbats_prophet",
    "var_pred": "w_var",
    "bilstm_pred": "w_bilstm",
}

REQUIRED_COLS = ["origin", "forecast_date", "horizon", "actual"] + MODEL_COLS

# ----------------------------------------------------------------------------
# 3. Utility functions
# ----------------------------------------------------------------------------
def assert_required_columns(df: pd.DataFrame, df_name: str) -> None:
    """Ensure the dataframe contains all required columns."""
    missing = [c for c in REQUIRED_COLS if c not in df.columns]
    if missing:
        raise ValueError(
            f"{df_name} is missing required columns: {missing}\n"
            f"Available columns: {list(df.columns)}"
        )

def safe_parse_dates(df: pd.DataFrame) -> pd.DataFrame:
    """Parse origin and forecast_date robustly."""
    out = df.copy()
    if "origin" in out.columns:
        out["origin"] = pd.to_datetime(out["origin"], errors="coerce")
    if "forecast_date" in out.columns:
        out["forecast_date"] = pd.to_datetime(out["forecast_date"], errors="coerce")
    return out

def prepare_input_df(df: pd.DataFrame, df_name: str) -> pd.DataFrame:
    """
    Prepare and validate a rolling-origin prediction dataframe.
    Applies:
      - schema checks
      - date parsing
      - numeric coercion
      - duplicate checks
      - chronological sort
    """
    out = df.copy()
    assert_required_columns(out, df_name)
    out = safe_parse_dates(out)

    # numeric coercion
    numeric_cols = ["horizon", "actual"] + MODEL_COLS
    for col in numeric_cols:
        out[col] = pd.to_numeric(out[col], errors="coerce")

    # drop rows that cannot be used
    out = out.dropna(subset=["origin", "forecast_date", "horizon", "actual"]).copy()
    out["horizon"] = out["horizon"].astype(int)

    # sort
    out = out.sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)

    # duplicate check on the forecast key
    key_cols = ["origin", "forecast_date", "horizon"]
    dup_count = out.duplicated(subset=key_cols).sum()
    if dup_count > 0:
        raise ValueError(
            f"{df_name} has {dup_count} duplicated forecast keys on {key_cols}. "
            "Fix duplicates before ensemble construction."
        )

    return out

def to_long_loss_table(df: pd.DataFrame, target_name: str) -> pd.DataFrame:
    """Convert wide prediction table to long model-wise loss table."""
    long_df = df.melt(
        id_vars=["origin", "forecast_date", "horizon", "actual"],
        value_vars=MODEL_COLS,
        var_name="model_col",
        value_name="prediction"
    ).copy()

    long_df["model"] = long_df["model_col"].map(MODEL_NAME_MAP)
    long_df["prediction"] = pd.to_numeric(long_df["prediction"], errors="coerce")
    long_df = long_df.dropna(subset=["prediction"]).copy()

    long_df["target"] = target_name
    long_df["error"] = long_df["actual"] - long_df["prediction"]
    long_df["abs_error"] = np.abs(long_df["error"])
    long_df["sq_error"] = long_df["error"] ** 2

    # APE with zero protection
    long_df["ape"] = np.where(
        np.abs(long_df["actual"]) > 1e-12,
        np.abs(long_df["error"]) / np.abs(long_df["actual"]) * 100,
        np.nan
    )

    # sMAPE component with zero protection
    denom = np.abs(long_df["actual"]) + np.abs(long_df["prediction"])
    long_df["smape_component"] = np.where(
        denom > 1e-12,
        2 * np.abs(long_df["prediction"] - long_df["actual"]) / denom * 100,
        np.nan
    )

    return long_df.sort_values(["origin", "horizon", "forecast_date", "model"]).reset_index(drop=True)

def rmse(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.sqrt(np.mean((y - yhat) ** 2))

def mae(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.mean(np.abs(y - yhat))

def mape(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    mask = np.abs(y) > 1e-12
    if mask.sum() == 0:
        return np.nan
    return np.mean(np.abs((y[mask] - yhat[mask]) / y[mask])) * 100

def smape(y, yhat):
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    denom = np.abs(y) + np.abs(yhat)
    mask = denom > 1e-12
    if mask.sum() == 0:
        return np.nan
    return np.mean(2 * np.abs(yhat[mask] - y[mask]) / denom[mask]) * 100

def mase_scale_from_unique_actuals(df: pd.DataFrame) -> float:
    """
    Compute a robust MASE denominator using unique forecast dates.
    This avoids double counting repeated actuals across rolling origins.
    """
    actual_series = (
        df[["forecast_date", "actual"]]
        .drop_duplicates()
        .sort_values("forecast_date")
        .copy()
    )

    if len(actual_series) < 2:
        return np.nan

    diffs = np.abs(np.diff(actual_series["actual"].values.astype(float)))
    diffs = diffs[~np.isnan(diffs)]

    if len(diffs) == 0:
        return np.nan

    scale = np.mean(diffs)
    if np.isclose(scale, 0):
        return np.nan

    return float(scale)

def mase(y, yhat, scale):
    if scale is None or np.isnan(scale) or np.isclose(scale, 0):
        return np.nan
    y = np.asarray(y, dtype=float)
    yhat = np.asarray(yhat, dtype=float)
    return np.mean(np.abs(y - yhat)) / scale

def compute_equal_weights() -> dict:
    """Equal weights fallback."""
    w = 1.0 / len(MODEL_COLS)
    return {m: w for m in MODEL_COLS}

def compute_inverse_rmse_weights(past_df: pd.DataFrame) -> dict:
    """
    Compute inverse-RMSE weights from a past-only sample.
    Expects columns: actual + base model predictions.
    """
    rmses = {}
    for model in MODEL_COLS:
        tmp = past_df[["actual", model]].dropna()
        if len(tmp) == 0:
            rmses[model] = np.nan
        else:
            rmses[model] = rmse(tmp["actual"], tmp[model])

    # Keep only finite positive RMSE values
    valid = {k: v for k, v in rmses.items() if pd.notna(v) and np.isfinite(v) and v > 0}

    # If not enough valid models, fall back to equal weights
    if len(valid) < 2:
        return compute_equal_weights()

    inv = {k: 1.0 / v for k, v in valid.items()}
    inv_sum = sum(inv.values())

    if inv_sum <= 0 or not np.isfinite(inv_sum):
        return compute_equal_weights()

    weights = {k: inv.get(k, 0.0) / inv_sum for k in MODEL_COLS}

    # Guard against numerical drift
    total = sum(weights.values())
    if total <= 0 or not np.isfinite(total):
        return compute_equal_weights()

    weights = {k: v / total for k, v in weights.items()}
    return weights

def build_dynamic_ensemble(df: pd.DataFrame, target_name: str, min_past_origins: int = 3) -> tuple[pd.DataFrame, pd.DataFrame]:
    """
    Build leakage-safe dynamic weighted ensemble.

    Weight rule:
    ------------
    For each current origin and horizon, use only rows with origin < current_origin.
    Prefer past rows from the same horizon. If there is insufficient same-horizon history,
    fall back to all past rows.

    Minimum history rule:
    ---------------------
    If the number of distinct past origins is less than min_past_origins,
    use equal weights.
    """
    out = df.copy().sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)

    for wcol in WEIGHT_COL_MAP.values():
        out[wcol] = np.nan

    out["ensemble_pred"] = np.nan
    weight_records = []

    unique_origins = sorted(out["origin"].dropna().unique())

    for current_origin in unique_origins:
        current_mask = out["origin"] == current_origin
        current_rows = out.loc[current_mask].copy()

        # past rows only -> leakage-safe
        past_rows_all = out.loc[out["origin"] < current_origin, ["origin", "horizon", "actual"] + MODEL_COLS].copy()
        n_past_origins = past_rows_all["origin"].nunique()

        for idx, row in current_rows.iterrows():
            current_horizon = int(row["horizon"])

            # default fallback
            weights = compute_equal_weights()
            weight_source = "equal_weights"

            if n_past_origins >= min_past_origins and len(past_rows_all) > 0:
                # prefer same-horizon past performance
                past_same_h = past_rows_all.loc[past_rows_all["horizon"] == current_horizon].copy()
                n_past_origins_same_h = past_same_h["origin"].nunique()

                if n_past_origins_same_h >= min_past_origins:
                    weights = compute_inverse_rmse_weights(past_same_h)
                    weight_source = "same_horizon_past_rmse"
                else:
                    weights = compute_inverse_rmse_weights(past_rows_all)
                    weight_source = "all_past_rmse"

            # ensemble prediction for current row
            pred = 0.0
            valid_weight_sum = 0.0

            for model in MODEL_COLS:
                model_value = row[model]
                model_weight = weights.get(model, 0.0)

                if pd.notna(model_value) and np.isfinite(model_value):
                    pred += model_weight * float(model_value)
                    valid_weight_sum += model_weight

                out.at[idx, WEIGHT_COL_MAP[model]] = model_weight

            # if some model predictions are missing, renormalize over available models
            if valid_weight_sum > 0 and not np.isclose(valid_weight_sum, 1.0):
                pred = pred / valid_weight_sum

            out.at[idx, "ensemble_pred"] = pred

            weight_records.append({
                "target": target_name,
                "origin": current_origin,
                "forecast_date": row["forecast_date"],
                "horizon": current_horizon,
                "weight_source": weight_source,
                "n_past_origins": int(n_past_origins),
                "w_sarimax": out.at[idx, "w_sarimax"],
                "w_tbats_prophet": out.at[idx, "w_tbats_prophet"],
                "w_var": out.at[idx, "w_var"],
                "w_bilstm": out.at[idx, "w_bilstm"],
            })

    weights_df = pd.DataFrame(weight_records).sort_values(["origin", "horizon", "forecast_date"]).reset_index(drop=True)
    return out, weights_df

def evaluate_models(df: pd.DataFrame, target_name: str) -> pd.DataFrame:
    """
    Evaluate base models plus ensemble using:
      - RMSE
      - MAE
      - MAPE
      - sMAPE
      - MASE
    """
    scale = mase_scale_from_unique_actuals(df)

    eval_models = MODEL_COLS + ["ensemble_pred"]
    results = []

    for model in eval_models:
        tmp = df[["actual", model]].dropna().copy()
        if len(tmp) == 0:
            continue

        y = tmp["actual"].values
        yhat = tmp[model].values

        results.append({
            "target": target_name,
            "model": MODEL_NAME_MAP.get(model, "Dynamic Weighted Ensemble"),
            "RMSE": rmse(y, yhat),
            "MAE": mae(y, yhat),
            "MAPE": mape(y, yhat),
            "sMAPE": smape(y, yhat),
            "MASE": mase(y, yhat, scale),
            "n_forecasts": len(tmp)
        })

    results_df = pd.DataFrame(results).sort_values(["target", "RMSE"]).reset_index(drop=True)
    return results_df

def plot_actual_vs_models(df: pd.DataFrame, target_name: str, save_dir: Path) -> None:
    """Plot actual vs ensemble vs best base model by RMSE."""
    eval_df = evaluate_models(df, target_name)

    base_only = eval_df[eval_df["model"] != "Dynamic Weighted Ensemble"].copy()
    if len(base_only) == 0:
        return

    best_base_model_name = base_only.iloc[0]["model"]
    reverse_name_map = {v: k for k, v in MODEL_NAME_MAP.items()}
    best_base_col = reverse_name_map[best_base_model_name]

    plot_df = df.sort_values("forecast_date").copy()

    plt.figure(figsize=(14, 6))
    plt.plot(plot_df["forecast_date"], plot_df["actual"], linewidth=2, label="Actual")
    plt.plot(plot_df["forecast_date"], plot_df["ensemble_pred"], linewidth=2, linestyle="--", label="Dynamic Weighted Ensemble")
    plt.plot(plot_df["forecast_date"], plot_df[best_base_col], linewidth=1.8, linestyle=":", label=f"Best Base Model ({best_base_model_name})")
    plt.title(f"{target_name.title()} — Actual vs Ensemble vs Best Base Model")
    plt.xlabel("Forecast Date")
    plt.ylabel("Value")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_actual_vs_ensemble_bestbase.png", dpi=300, bbox_inches="tight")
    plt.show()

def plot_average_weights_by_origin(weights_df: pd.DataFrame, target_name: str, save_dir: Path) -> None:
    """Plot average weights across origins."""
    if weights_df.empty:
        return

    origin_avg = (
        weights_df.groupby("origin")[["w_sarimax", "w_tbats_prophet", "w_var", "w_bilstm"]]
        .mean()
        .reset_index()
        .sort_values("origin")
    )

    plt.figure(figsize=(14, 6))
    plt.plot(origin_avg["origin"], origin_avg["w_sarimax"], label="SARIMAX")
    plt.plot(origin_avg["origin"], origin_avg["w_tbats_prophet"], label="TBATS-Prophet")
    plt.plot(origin_avg["origin"], origin_avg["w_var"], label="VAR")
    plt.plot(origin_avg["origin"], origin_avg["w_bilstm"], label="BiLSTM-Attention")
    plt.title(f"{target_name.title()} — Average Model Weights by Origin")
    plt.xlabel("Origin")
    plt.ylabel("Average Weight")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_weights_by_origin.png", dpi=300, bbox_inches="tight")
    plt.show()

def plot_average_weights_by_horizon(weights_df: pd.DataFrame, target_name: str, save_dir: Path) -> None:
    """Plot average weights by horizon."""
    if weights_df.empty:
        return

    horizon_avg = (
        weights_df.groupby("horizon")[["w_sarimax", "w_tbats_prophet", "w_var", "w_bilstm"]]
        .mean()
        .reset_index()
        .sort_values("horizon")
    )

    plt.figure(figsize=(14, 6))
    plt.plot(horizon_avg["horizon"], horizon_avg["w_sarimax"], marker="o", label="SARIMAX")
    plt.plot(horizon_avg["horizon"], horizon_avg["w_tbats_prophet"], marker="o", label="TBATS-Prophet")
    plt.plot(horizon_avg["horizon"], horizon_avg["w_var"], marker="o", label="VAR")
    plt.plot(horizon_avg["horizon"], horizon_avg["w_bilstm"], marker="o", label="BiLSTM-Attention")
    plt.title(f"{target_name.title()} — Average Model Weights by Horizon")
    plt.xlabel("Forecast Horizon")
    plt.ylabel("Average Weight")
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig(save_dir / f"{target_name}_weights_by_horizon.png", dpi=300, bbox_inches="tight")
    plt.show()

# ----------------------------------------------------------------------------
# 4. Load and validate rolling-origin prediction files
# ----------------------------------------------------------------------------
print("=" * 90)
print("STEP 1 — LOADING AND VALIDATING INPUT FILES")
print("=" * 90)

data_raw = {}
data_prepared = {}
loss_tables = {}

for target, path in FILES.items():
    if not path.exists():
        raise FileNotFoundError(f"Missing required file: {path}")

    df = pd.read_csv(path)
    data_raw[target] = df
    data_prepared[target] = prepare_input_df(df, target)
    loss_tables[target] = to_long_loss_table(data_prepared[target], target)

    print(f"\n[{target.upper()}]")
    print(f"File: {path}")
    print(f"Rows loaded: {len(df):,}")
    print(f"Rows after preparation: {len(data_prepared[target]):,}")
    print(f"Unique origins: {data_prepared[target]['origin'].nunique():,}")
    print(f"Unique forecast dates: {data_prepared[target]['forecast_date'].nunique():,}")
    print(f"Horizon range: {data_prepared[target]['horizon'].min()} to {data_prepared[target]['horizon'].max()}")

print("\nSchema validation completed successfully.")

# ----------------------------------------------------------------------------
# 5. Show long-format loss tables
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 2 — LONG-FORMAT MODEL-WISE LOSS TABLES")
print("=" * 90)

for target, long_df in loss_tables.items():
    print(f"\nLong loss table preview — {target}")
    display(long_df.head())

# ----------------------------------------------------------------------------
# 6. Build dynamic weighted ensemble for each target
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 3 — BUILDING DYNAMIC WEIGHTED ENSEMBLES")
print("=" * 90)

ensemble_data = {}
weight_logs = {}

for target, df in data_prepared.items():
    ens_df, w_df = build_dynamic_ensemble(df, target_name=target, min_past_origins=3)
    ensemble_data[target] = ens_df
    weight_logs[target] = w_df

    print(f"\n[{target.upper()}] Dynamic ensemble created")
    print(f"Rows: {len(ens_df):,}")
    print(f"Weight log rows: {len(w_df):,}")
    print("Weight source counts:")
    print(w_df["weight_source"].value_counts(dropna=False))

# ----------------------------------------------------------------------------
# 7. Evaluate base models vs ensemble
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 4 — PERFORMANCE EVALUATION")
print("=" * 90)

metrics_tables = []

for target, df in ensemble_data.items():
    result = evaluate_models(df, target)
    metrics_tables.append(result)

    print(f"\nPerformance comparison — {target}")
    display(result)

metrics_all = pd.concat(metrics_tables, axis=0, ignore_index=True)
print("\nCombined performance table")
display(metrics_all.sort_values(["target", "RMSE"]).reset_index(drop=True))

# ----------------------------------------------------------------------------
# 8. Diagnostics plots
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 5 — DIAGNOSTIC PLOTS")
print("=" * 90)

for target in ["unemployment", "vacancies", "theta"]:
    print(f"\nGenerating plots for {target} ...")
    plot_actual_vs_models(ensemble_data[target], target, FIG_DIR)
    plot_average_weights_by_origin(weight_logs[target], target, FIG_DIR)
    plot_average_weights_by_horizon(weight_logs[target], target, FIG_DIR)

# ----------------------------------------------------------------------------
# 9. Export ensemble outputs for RF residual modeling
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("STEP 6 — EXPORTING ENSEMBLE FILES")
print("=" * 90)

export_files = {
    "unemployment": ENSEMBLE_DIR / "ensemble_unemployment.csv",
    "vacancies": ENSEMBLE_DIR / "ensemble_vacancies.csv",
    "theta": ENSEMBLE_DIR / "ensemble_theta.csv",
}

for target, out_path in export_files.items():
    export_cols = [
        "origin", "forecast_date", "horizon", "actual",
        "sarimax_pred", "tbats_prophet_pred", "var_pred", "bilstm_pred",
        "w_sarimax", "w_tbats_prophet", "w_var", "w_bilstm",
        "ensemble_pred"
    ]
    ensemble_data[target][export_cols].to_csv(out_path, index=False)
    print(f"Exported: {out_path}")

# optional: export weight logs too
for target, wdf in weight_logs.items():
    wdf.to_csv(ENSEMBLE_DIR / f"ensemble_weight_log_{target}.csv", index=False)

# optional: export metrics table
metrics_all.to_csv(ENSEMBLE_DIR / "ensemble_metrics_summary.csv", index=False)

# ----------------------------------------------------------------------------
# 10. Final audit summary
# ----------------------------------------------------------------------------
print("\n" + "=" * 90)
print("FINAL AUDIT SUMMARY")
print("=" * 90)

for target in ["unemployment", "vacancies", "theta"]:
    df = ensemble_data[target]
    print(f"\n[{target.upper()}]")
    print(f"Export rows: {len(df):,}")
    print(f"Missing ensemble predictions: {df['ensemble_pred'].isna().sum():,}")
    print("Average weights:")
    print(df[["w_sarimax", "w_tbats_prophet", "w_var", "w_bilstm"]].mean().round(4))

print("\nDynamic weighted ensemble pipeline completed successfully.")
print("Next notebook: RF residual hybrid using the exported ensemble files.")
```
